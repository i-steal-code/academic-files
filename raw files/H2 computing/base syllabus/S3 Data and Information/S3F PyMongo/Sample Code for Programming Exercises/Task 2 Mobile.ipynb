```python
import pymongo, json
client = pymongo.MongoClient("localhost",27017)

coll = client["jp_mobile"]["phone"]
x = coll.delete_many({})   # Remove all the exisiting documents in collection
print(x.deleted_count,"documents deleted.")

with open("items.json") as file:
    data = json.load(file)
file.close() #file is closed
coll.insert_many(data)
#client.close()
```

```python
while True:
    brand = input("Enter brand: ")
    model = input("Enter model: ")
    color = input("Enter colour: ")
    price = int(input("Enter price: "))
    qty   = int(input("Enter quantity: "))
    gift = []
    noOfGift = int(input("How many free gifts: "))
    for _ in range(noOfGift):
        freeGift = input("Enter free gift: ")
        gift.append(freeGift)
    
    # Search for existing phone and add quantity if found
    search = {"brand":brand, "model":model, "colour":color}
    if coll.count_documents(search) > 0:
        update_data = {
            "$set": {"price":price, "gift":gift},
            "$inc": {"quantity": qty}
        }
        coll.update_one(search, update_data)
    else:
        new_doc = {"brand":brand, "model":model, "colour":color, "price":price, "quantity":qty, "gift":gift}
        coll.insert_one(new_doc)
        
    more = input("Any more phone to input [Y/N]? ")
    if more[0].upper() != 'Y':
        break
    
#client.close()
```

```python
def display_all():
    result = coll.find()
    print("All documents in collection:")
    print("{:<10}{:<10}{:<10}{:<10}{:<10}{:<30}".format("brand","model","colour","price","quantity","free gifts"))
    for doc in result:
        gifts_list = doc.get("gift",[])
        if len(gifts_list)>0:
            gifts_str = ", ".join(gifts_list)
        else:
            gifts_str = "None"
        print("{:<10}{:<10}{:<10}{:<10}{:<10}{:<30}".format(doc.get("brand"),\
                                                           doc.get("model"),\
                                                           doc.get("colour"),\
                                                           doc.get("price"),\
                                                           doc.get("quantity"),\
                                                            gifts_str))
```

```python
display_all()
client.close()
```

```python
import sqlite3
conn = sqlite3.connect("school.db") #connect to database

#1. Create a new table named "course" with columns: 
#course_id (integer, primary key), course_name (text), credit (real).
conn.execute("DROP TABLE course")
conn.execute("""
                CREATE TABLE IF NOT EXISTS course(
                course_id INTEGER PRIMARY KEY, 
                course_name TEXT, credit REAL
                )
             """)

#2. Get user input for 5 courses and insert them into the "course" table.
for _ in range(5):
    course_id = int(input("Enter course ID: "))
    course_name = input("Enter course name: ")
    credit = float(input("Enter course credit (2.0 to 4.0): "))
    conn.execute("""INSERT INTO course (course_id,course_name,credit)
                   VALUES (?,?,?)""", (course_id,course_name,credit))
    conn.commit()
    
#3. Write a query to retrieve all courses with more than 3 credits.
cursor = conn.execute("SELECT * FROM course WHERE credit > 3.0")
rows = cursor.fetchall()
print("\nCourses with more than 3 credits:")
for row in rows:
    print(row)
    
#4. Write a query to retrieve all courses sorted by their names.
cursor = conn.execute("SELECT * FROM course ORDER BY course_name")
sorted_courses = cursor.fetchall()
print("\nCourses sorted by name:")
for row in sorted_courses:
    print(row)

#5. Write a query to find the average number of credits for all courses.
cursor = conn.execute("SELECT AVG(credit) FROM course")
print("Average number of credits for all courses:",cursor.fetchone()[0])

conn.close() #close database
```

```python
import sqlite3
conn = sqlite3.connect("school.db") #connect to database

#1. Create a new table named "enrolment" with columns: 
#enrolment_id (integer), student_id (integer), 
#course_id (integer), enrolment_date (text).
conn.execute("DROP TABLE enrolment")
conn.execute("""CREATE TABLE IF NOT EXISTS enrolment (
                enrolment_id INTEGER PRIMARY KEY,
                student_id INTEGER,
                course_id INTEGER,
                enrolment_date TEXT)""")

#2. Insert several records into the "enrolment" table, indicating the courses in which the students are enrolled.
conn.execute("INSERT INTO enrolment VALUES (1,1,11,'2026-01-01')")
conn.execute("INSERT INTO enrolment VALUES (2,2,11,'2026-01-01')")
conn.execute("INSERT INTO enrolment VALUES (3,3,11,'2026-01-01')")
conn.commit()

#3. Write a query to find all students enrolled in a specific course.
requested_course_id = int(input("Enter course id to find students: "))
cursor = conn.execute("""SELECT Student.name FROM Student
                INNER JOIN enrolment ON Student.id = enrolment.student_id
                WHERE enrolment.course_id = ?""", (requested_course_id,))
enrolled_students = cursor.fetchall()
print(f"\nStudents enrolled in course {requested_course_id}: ")
for row in enrolled_students:
    print(row[0])

#4. Write a query to find all courses a specific student is enrolled in.
requested_student_id = int(input("Enter student id to find course enrolled in: "))
cursor = conn.execute("""SELECT course.course_name FROM course
                        INNER JOIN enrolment 
                        ON enrolment.course_id = course.course_id
                        WHERE enrolment.student_id = ?""",(requested_student_id,))
enrolled_in_courses = cursor.fetchall()
print(f"\nCourses enrolled by student {requested_student_id}:")
for row in enrolled_in_courses:
    print(row[0])

#5. Update the enrolment date for a specific enrolment record.
update_enrolment_id = int(input("Enter enrolment id to update:"))
new_enrol_date = input("Enter new enrolment date: ")
cursor = conn.execute("""UPDATE enrolment SET enrolment_date=?
                      WHERE enrolment_id=?
                      """,(new_enrol_date,update_enrolment_id))
print("Enrolment date updated.")

#6. Write a query to find the total number of students enrolled in each course.
cursor = conn.execute("""SELECT course.course_name, COUNT(enrolment.student_id) AS Total_students
                        FROM course
                        LEFT OUTER JOIN enrolment ON course.course_id = enrolment.course_id
                        GROUP BY course.course_id""")
rows = cursor.fetchall()
for row in rows:
    print(f"{row[1]} students enrolled in {row[0]} course")

#7. Write a query to display the student names along with the courses they are enrolled in. Display all students even if they have no courses enrolled.
print("\nALL STUDENTS AND THEIR COURSES")
cursor = conn.execute("""SELECT student.name, course.course_name 
            FROM student 
            LEFT OUTER JOIN enrolment ON student.id = enrolment.student_id
            LEFT OUTER JOIN course ON enrolment.course_id = course.course_id""")
for row in cursor.fetchall():
    print(f"{row[0]}: {row[1]}")

conn.close() #close database
```

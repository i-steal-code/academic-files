```python
import sqlite3
while True:
    conn = sqlite3.connect("loans2B.db")
    # Get Borrower ID
    borrower_id = int(input("Enter Borrower ID: "))
    cursor = conn.execute("SELECT COUNT(*) FROM Borrower WHERE ID = ?", (borrower_id,))
    if cursor.fetchone()[0] == 0:
        name = input("Enter Borrower Name: ")
        conn.execute("INSERT INTO Borrower(ID, Name) VALUES(?, ?)", (borrower_id, name))
    # Get BookID
    book_id = int(input("Enter Book ID: "))
    cursor = conn.execute("SELECT COUNT(*) FROM Book WHERE ID = ?", (book_id,))
    if cursor.fetchone()[0] == 0:
        title = input("Enter book title: ")
        conn.execute("INSERT INTO Book(ID, Title) VALUES (?,?)", (book_id, title))
    # Insert loan
    cursor = conn.execute("SELECT COUNT(*) FROM Loan WHERE BookID = ?", (book_id,))
    if cursor.fetchone()[0] == 0: #not borrowed because it's zero
        conn.execute("INSERT INTO Loan(BorrowerID, BookID) VALUES (?,?)", (borrower_id, book_id))
        conn.commit()
    else: # book is already loaned out because loan record for that book is not zero
        print("Error: Book is already on loan")
        conn.rollback()
    conn.close()
    quit = input("Quit (Y/N)?: ")
    if quit.upper() == 'Y':
        break
```

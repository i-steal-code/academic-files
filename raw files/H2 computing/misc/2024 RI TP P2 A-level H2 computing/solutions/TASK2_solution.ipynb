```python
#Task 2.1
import sqlite3

# Connect to the database (or create it if it doesn't exist)
conn = sqlite3.connect('grocery.db')
cursor = conn.cursor()

cursor.execute('''DROP TABLE IF EXISTS products''')
cursor.execute('''DROP TABLE IF EXISTS suppliers''')
cursor.execute('''DROP TABLE IF EXISTS transactions''')

# Create product table
cursor.execute('''CREATE TABLE IF NOT EXISTS products (
                    ProductID INTEGER PRIMARY KEY,
                    ProductName TEXT NOT NULL,
                    Category TEXT,
                    UnitPrice REAL,
                    StockQuantity INTEGER,
                    SupplierID INTEGER,
                    FOREIGN KEY (SupplierID) REFERENCES supplier(SupplierID)
                )''')

# Create supplier table
cursor.execute('''CREATE TABLE IF NOT EXISTS suppliers (
                    SupplierID INTEGER PRIMARY KEY,
                    SupplierName TEXT NOT NULL,
                    ContactNumber TEXT
                )''')

# Create transaction table
cursor.execute('''CREATE TABLE IF NOT EXISTS transactions (
                    TransactionID INTEGER PRIMARY KEY,
                    ProductID INTEGER,
                    Quantity INTEGER,
                    TransactionType TEXT,
                    TransactionDate TEXT,
                    FOREIGN KEY (ProductID) REFERENCES product(ProductID)
                )''')

# Commit changes and close connection
conn.commit()
conn.close()
```

```python
#Task 2.3
import sqlite3

def display_products_by_category(category):
    conn = sqlite3.connect('grocery.db')
    cursor = conn.cursor()
    cursor.execute('''SELECT p.ProductID, p.ProductName, p.UnitPrice, p.SupplierID, s.SupplierName 
                      FROM products p 
                      JOIN suppliers s ON p.SupplierID = s.SupplierID 
                      WHERE p.Category = ?''', (category,))
    products = cursor.fetchall()
    conn.close()
    return products

category = input("Enter the category of the product: ")
products = display_products_by_category(category)
for product in products:
    print(product)
```

```python
#Task 1.1
class EmployeeNode:
    def __init__(self, emp_id, name, salary):
        self.__emp_id = emp_id
        self.__name = name
        self.__salary = salary
        self.__next = None

    def get_emp_id(self):
        return self.__emp_id

    def set_emp_id(self, emp_id):
        self.__emp_id = emp_id

    def get_name(self):
        return self.__name

    def set_name(self, name):
        self.__name = name

    def get_salary(self):
        return self.__salary

    def set_salary(self, salary):
        self.__salary = salary

    def get_next(self):
        return self.__next

    def set_next(self, next_node):
        self.__next = next_node
```

```python
#Task 1.2
class EmployeeLinkedList:
    def __init__(self):
        self.__head = None

    def AddEmployee(self, emp_id=None, name=None, salary=None):
        if emp_id is None:
            emp_id = input("Enter employee ID: ")
        if name is None:
            name = input("Enter employee name: ")
        if salary is None:
            salary = input("Enter employee salary: ")

        new_employee = EmployeeNode(emp_id, name, salary)

        if self.__head is None: #add to empty linked list
            self.__head = new_employee
        elif emp_id < self.__head.get_emp_id(): #add as first node
            new_employee.set_next(self.__head)
            self.__head = new_employee
        else: #add as non-first node
            current = self.__head
            while current.get_next() is not None and current.get_next().get_emp_id() < emp_id:
                current = current.get_next()
            new_employee.set_next(current.get_next())
            current.set_next(new_employee)

    def DisplayAll(self, current=None):
        if current is None:
            current = self.__head
        if current is not None:
            print(f"Employee ID: {current.get_emp_id()}, Name: {current.get_name()}, Salary: {current.get_salary()}")
            if current.get_next() is not None:
                self.DisplayAll(current.get_next())
            
    def RemoveEmployee(self, emp_id=None):
        if emp_id is None:
            emp_id = input("Enter employee ID to remove: ")

        current = self.__head
        previous = None
        while current is not None and current.get_emp_id() != emp_id:
            previous = current
            current = current.get_next()

        if current is not None:
            if previous is None:
                self.__head = current.get_next()
            else:
                previous.set_next(current.get_next())
        else:
            print("Employee not found.")

    def UpdateSalary(self, emp_id=None, salary=None):
        if emp_id is None:
            emp_id = input("Enter employee ID to update salary: ")
        if salary is None:
            salary = input("Enter new salary: ")

        current = self.__head
        while current is not None and current.get_emp_id() != emp_id:
            current = current.get_next()

        if current is not None:
            current.set_salary(salary)
        else:
            print("Employee not found.")
```

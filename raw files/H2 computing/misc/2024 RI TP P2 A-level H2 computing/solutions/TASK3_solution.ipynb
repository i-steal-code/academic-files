```python
import socket

# Define the client class
class TreasureHuntClient:
    def __init__(self):
        self.socket = socket.socket()   
        self.socket.connect(('localhost', 5000)) #1m

    def start_game(self):
        # Receive the initial state of the map from the server
        print("Welcome to Treasure Hunt!")
        print("Use 'up', 'down', 'left', 'right' to navigate.")
        print("Find the treasures ('T') hidden in the map.") #1m
        self.display_map()

        # Start the game loop
        self.play_game()
        # Close the socket
        self.socket.close()  #1m

    def display_map(self):
        # Receive and display the current state of the map from the server
        map_data = self.socket.recv(1024).decode() #1m
        print(map_data)  #1m

    def play_game(self):
        # Game loop
        while True:  #1m
            
            # Prompt the player to enter their move
            move = input("Enter your move (up/down/left/right): ").strip().lower() #1m

            # Send the player's move to the server
            self.socket.sendall(move.encode())  #1m
            data = self.socket.recv(1024).decode()  #1m
            if data == "You found a treasure!\n":
                print(data)  #1m
                data = self.socket.recv(1024).decode()
                if data == "Congratulations! You found all treasures.\n":  #1m
                    print(data)
                    return  #1m
                else:
                    print(data)
            else:
                print(data)  #1m

# Main function to start the client
def main():
    client = TreasureHuntClient()  #1m
    client.start_game()

if __name__ == "__main__":
    main()
```

```python
#Task 1.1
def encode_str(text, token):

    if token in text: #validate input, ensure "token" is not in "text"
        return f"Error: text cannot contain the token {token}."
    
    for char in text:
        if char.isdigit():
            return f"Error: Text cannot contain digits."
        
    i=0
    result = []
    
    while i < len(text):
        curr_char = text[i]
        count = 1
        
        while i+count < len(text) and text[i+count] == curr_char: #count consecutive characters
            count+=1
        
        if count >= 3:
            result.append(f"{token}{curr_char}{count}")
        else:
            result.append(curr_char * count)
        
        i += count
    return ''.join(result)

print(encode_str("aaabbcc","%"))  #output %a3bbcc
print(encode_str("aabbbccc","%"))  #output aa%b3%c3
print(encode_str("aabbbccccccccccccaaaaaaaaccccccc","&"))  #output aa&b3&c14&a8&c7
print(encode_str("aaab%bcc","%"))  #error cannot contain token
print(encode_str("aaa3bbcc","%"))  #error cannot contain digits
```

```python
#Task 1.2
def decode_str(encoded, token):
    i=0
    result = []
    
    while i < len(encoded):
        if encoded[i] == token: #token found
            char = encoded[i+1]
            count = '' #extract count
            j = i+2
            
            while j < len(encoded) and encoded[j].isdigit():
                count += encoded[j]
                j += 1
            
            result.append(char*int(count))
            i = j
        else:
            result.append(encoded[i])
            i += 1
    return "".join(result)

#testing code
print(decode_str("aa&b3&c31","&"))
```

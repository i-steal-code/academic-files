```python
#Task 2.1

import random

def generate_runners():
    phone_nums = []
    categories = ["Men's Marathon","Men's Half-Marathon","Women's Marathon","Women's Half-Marathon"]
    runners = []
    
    while len(phone_nums) < 15000:  #generate 15000 distinct 8-digit numbers
        num = random.randint(80000000, 98999999)
        if num not in phone_nums:
            phone_nums.append(num)
    
    for num in phone_nums: #pair numbers with category randomly
        category = random.choice(categories)
        runners.append((num,category))
    
    file = open("runners.txt",'w')
    for num,cat in runners:
        file.write(f"{num},{cat}\n")
    file.close()
    
    print("Generated",len(runners),"runners and saved file.")
    return runners

#Test code
runners = generate_runners()
```

```python
#Task 2.2

def merge_sort(arr):
    if len(arr) <=1:
        return arr
    mid = len(arr)//2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])
    return merge(left,right)

def merge(left,right):
    result = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i][0] < right[j][0]: #phone numbers is in index 0
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    result = result + left[i:] + right[j:]
    return result

#Measure time taken using timeit module
import timeit
start_time = timeit.default_timer()
sorted_data = merge_sort(runners)
end_time = timeit.default_timer()
time_difference = end_time - start_time
print(time_difference)

file = open("runners_sorted_1.txt",'w')
for num,cat in sorted_data:
    file.write(f"{num},{cat}\n")
file.close()
```

```python
#Task 2.3
class Node:
    
    def __init__(self, phone, cat):
        self.phone = phone
        self.category = cat
        self.next = None
    
    def getPhone(self):
        return self.phone
    
    def getCategory(self):
        return self.category
    
    def getNext(self):
        return self.next
    
    def setNext(self, ptr):
        self.next = ptr

class OrderedLL:
    
    def __init__(self):
        self.head = None
        
    def insert(self, phone, cat):
        new_node = Node(phone, cat)
        
        if self.head == None or phone < self.head.getPhone(): #insert into empty LL or first node
            new_node.setNext(self.head)
            self.head = new_node
        else:
            current = self.head
            while current.getNext() is not None and current.getPhone() < phone:
                current = current.getNext()
                
            new_node.setNext(current.getNext())
            current.setNext(new_node)
    
    def delete(self, phone):
        if self.head is None:
            return False
        
        if self.head.getPhone() == phone:
            self.head = self.head.getNext()
            return True
        
        current = self.head
        while current.getNext() is not None and current.getNext().getPhone != phone:
            current = current.getNext()
            
        if current.getNext() is not None:
            current.setNext(current.getNext().getNext())
            return True
        
        return False
    
    def update(self, phone, new_cat):
        current = self.head
        while current:
            if current.getPhone == phone:
                current.setCategory(new_cat)
                return True
            current = current.getNext()
        return False
    
    def get_list(self):
        result = []
        current = self.head
        while current:
            result.append((current.getPhone(), current.getCategory()))
            current = current.getNext()
        return result
    
#Main
runners_LL = OrderedLL()

start_time = timeit.default_timer()
for phone,cat in sorted_data:
    runners_LL.insert(phone,cat)
# print(runners_LL.get_list())
end_time = timeit.default_timer()
time_difference = end_time - start_time
print(time_difference)

sorted_OLL = runners_LL.get_list()
file = open("runners_sorted_1.txt",'w')
for num,cat in sorted_OLL:
    file.write(f"{num},{cat}\n")
file.close()
```

```python
#Task 2.4

def bin_search(arr, target, left, right):
    if left > right:
        return "Runner not found"
    
    mid = (left + right) // 2
    mid_val = arr[mid][0] #extract phone number at middle index
    if target == mid_val:
        return arr[mid][1]  #Found match, return category
    elif target < mid_val:
        return bin_search(arr, target, left, mid-1)
    else:
        return bin_search(arr, target, mid+1, right)
    
#testing code
test1 = sorted_data[len(sorted_data)-1][0]  #last phone number
test2 = sorted_data[len(sorted_data)//2][0]  #middle phone number
test3 = 99999999 #number not in list

for test in (test1, test2, test3):
    result = bin_search(sorted_data, test, 0, len(sorted_data)-1)
    print(f"Search for {test}: {result}")
```

```python
#Task 2.5

class cir_Queue:
    
    def __init__(self, capacity):
        self.data = [None] * capacity
        self.front = 0
        self.rear = 0
        self.size = 0 #current number of items in queue
        self.capacity = capacity
        
    def is_empty(self):
        return self.size == 0
    
    def is_full(self):
        return self.size == self.capacity
    
    def enqueue(self, phone, category):
        if self.is_full():
            print("Queue is full. Cannot enqueue.")
            return
        
        self.data[self.rear] = (phone, category)
        self.rear = (self.rear + 1) % self.capacity #update rear pointer
        self.size += 1
        print("Enqueued:",phone,category)
        
    def dequeue(self):
        if self.is_empty():
            print("Queue is empty. Cannot dequeue.")
            return
        
        data = self.data[self.front]
        self.data[self.front] = None #clear data in front of queue, optional step
        self.front = (self.front+1) % self.capacity
        self.size -= 1
        print("Dequeued:",data)
        return data
    
    def display(self):
        if self.is_empty():
            print("Queue is empty")
            return
    
        print("Contents [front to rear]:")
        current = self.front
        for i in range(self.size):
            print(self.data[current])
            current = (current+1) % self.capacity

#Testing code
rq = cir_Queue(5)
print("Queue is empty", rq.is_empty())
rq.enqueue(80000000,"Men's Marathon")
rq.enqueue(80000001,"Men's Marathon")
rq.enqueue(80000002,"Men's Marathon")
rq.enqueue(80000003,"Men's Marathon")
rq.enqueue(80000004,"Women's Marathon")
rq.enqueue(80000005,"Women's Marathon")
print("Queue is full", rq.is_full())
rq.dequeue()
rq.dequeue()
rq.display()
```

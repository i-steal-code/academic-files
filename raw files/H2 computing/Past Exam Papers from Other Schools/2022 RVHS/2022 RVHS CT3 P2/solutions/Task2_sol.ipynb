```python
from pymongo import MongoClient

# Task 1
# Create a mongo client to local host with port 27017 [1]
# Create database "supermarket" with one collection "items" [1]
client = MongoClient('localhost', 27017)
client.drop_database('supermarket')
mydb = client['supermarket']
mycol = client['supermarket']['items']

# Put information in 'supermarket_items.txt' into the NoSQL database "supermarket" inside collection "items" 
# open file [1]
# readfile till eof [1]
# split lines and remove line character [2]
# convert type [1]
# create dic for insert [2]
# insert info to database [1]
f = open('supermarket_items.txt', 'r')
line = f.readline()
line = f.readline()
while line:
    item_name, item_type, price = line[:-1].split(',')
    mycol.insert_one({"item_name":item_name,"item_type":item_type,"price":float(price)})
    line = f.readline()
f.close()


# Task 2.2
# Menu flow with correct display [2]
# validation [1]
# opt1 [2]
# opt2 [2]
# opt3 [2]
# opt4 [1]
done = False
while not done:
    print("Choose an option:")
    print("1. Find cost by item name.")
    print("2. Find items by type.")
    print("3. Update cost by item name.")
    print("4. Quit")
    opt = input("Choose an option:")
    while len(opt)!=1 and opt not in "1234":
        opt = input("Choose an option again:")
    if opt == "1":
        item_name = input("Type an item name:")
        myquery = {"item_name":item_name}
        display = {"price":1}
        posts = list(mycol.find(myquery, display))
        if posts:
            print(item_name + " each cost $" + str(posts[0]["price"]) + ".")
        else:
            print("Item not found.")
            
    elif opt == "2":
        item_type = input("Type an item type:")
        myquery = {"item_type":item_type}
        display = {"item_name":1}
        posts = list(mycol.find(myquery, display))
        for i in range(len(posts)):
            print(str(i+1) + " " + posts[i]["item_name"])
            
    elif opt == "3":
        item_name = input("Type an item name you want to update price:")
        price = input("Type a new update price:")
        myquery = {"item_name":item_name}
        updates = {"$set":{"price":float(price)}}
        mycol.update_one(myquery, updates)
    elif opt == "4":
        done = True
        print("Program quitted.")
    
client.close()



```

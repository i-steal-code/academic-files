```python
#Task 1 menu code provided. Do not modify.

found_list = ['bottle', 'iphone', 'calculator', 'wallet', 'watch']

def menu():
    inputstring = "Default"
    while inputstring != "":
        print("\nMenu:")
        print("1. Check for Lost and Found items")
        print("2. Delete an item returned to student")
        print("3. Add a new Lost and Found item")
        inputstring = input("Select an option or press Enter to quit: ")
        if inputstring == "1":
            check()
        elif inputstring == "2":
            delete()
        elif inputstring == "3":
            add()
```

```python
# Task 1.1 answer (3 marks)
def check():
    keyword = input("Please type the keyword to check: ") # 1m input
    if keyword in found_list:                             # 1m if
        print("The item exists in the records.")          # 1m correct print
    else:
        print("The item does not exist in the records.")
```

```python
# Task 1.2 answer (3 marks)
def delete():
    keyword = input("Please type the keyword for the item to deleted: ")
    if keyword in found_list:                            # 1m if
        found_list.remove(keyword)                       # 1m delete/mutate
        print("The item has been deleted successfully.")
    else:
        print("The item does not exist in the records.") # 1m correct printout
```

```python
# Task 2.1                      (3 marks)
def hashkey(name):
    country = name.lower()      # 1 mark 
    total = 0
    for char in country:
        total += ord(char)      # 1 mark for calculating total
    return total % 30           # 1 mark  


hashkey('Brazil')
```

```python
# Task 2.2                        (6 marks) changed to 7
size = 30
HashTable = [''] * size                   #[1] initialize array size 30
count = 0
with open('COUNTRY1.TXT', 'r') as f:      #[1] read from data file 
    for line in f:
        country = line.strip()            #[1] iterate each country name (added this)
        address = hashkey(country)        #[1] compute hash address
        if HashTable[address] == '':      #[1] check empty  
            HashTable[address] = country  #[1] insert into hash table
            count += 1
        else:
            print("Collision:", country, "is not added.")  #[1] Error msg for collision

print(count)
print(HashTable)
```

```python
# Task 2.3                                (5 marks)
count = 0
with open('COUNTRY2.txt', 'r') as f:
    for line in f:
        country = line.strip()
        address = hashkey(country)
        if HashTable[address] == '':
            HashTable[address] = country
        else:
            index = address
            found = False

            while not found:                #[1] repeat until found  
                index += 1                  #[1] start from index or index+1
                if index == len(HashTable): #[1] loop back to front of array
                    index = 0
                
                if HashTable[index] == '':  #[1] check for empty before insert into hash table
                    HashTable[index] = country
                    found = True            #[1] break from while-loop
                    print(country)
                    count += 1
print(count)
print(HashTable)
```

```python
#Task 3.1 (7 marks)

class Package:
    def __init__(self, data, next = None):
        self.data = data
        self.next = next                                        
    
    def get_data(self):
        return self.data
    
    def get_next(self):                                        
        return self.next

    def set_data(self, new_data):
        self.data = new_data

    def set_next(self, new_next):                               
        self.next = new_next

class Queue:
    def __init__(self, root = None):
        self.root = root

    def IsEmpty(self):
        return self.root == None                                 

    def enqueue(self, data):
        package = Package(data, self.root)                       
        self.root = package                                      

    def dequeue(self):
        if self.IsEmpty():
            print('Empty Q')
            return False                                 #[1] check for empty Q
        else:
            current = self.root
            prev = None                                  #[1] assign temp var
            while current.get_next():
                prev = current
                current = current.get_next()             #[1] correct while-loop and advance
            temp = current.get_data()
            if prev == None:
                self.root = None                         #[1] check if first node
            else:
                prev.set_next(None)
            return temp
    
    def display(self):
        if self.IsEmpty():
            print('Empty Q')
        else:
            print('[tail]', end =' ')
            current = self.root                           #[1] assign temp
            while current.get_next():
                print(current.get_data(), end = ' -> ')
                current = current.get_next()              #[1] while-loop and advance
            print(current.get_data(), end = ' [head]\n')  #[1] correct print format
            

```

```python
#Task 3.3 (6 marks)

class Queue:
    def __init__(self, root = None):
        self.root = root

    def IsEmpty(self):
        return self.root == None

    def enqueue(self, data):
        package = Package(data, self.root)
        self.root = package

    def dequeue(self):
        if self.IsEmpty():
            print('Empty Q')
            return False
        else:
            current = self.root
            prev = None
            while current.get_next():
                prev = current
                current = current.get_next()
            temp = current.get_data()
            if prev == None:
                self.root = None
            else:
                prev.set_next(None)
            return temp
    
    def display(self):
        if self.IsEmpty():
            print('Empty Q')
        else:
            print('[tail]', end =' ')
            current = self.root
            while current.get_next():
                print(current.get_data(), end = ' -> ')
                current = current.get_next()
            print(current.get_data(), end = ' [head]\n')

    def cancel(self, data):
        if self.IsEmpty():
            print('Empty Q')
            return False                                    #[1] check for empty Q
        else:
            current = self.root
            prev = None                                     #[1] assign temp vars

            while current:
                if current.get_data() == data:              #[1] if data in current's data
                    if prev == None:
                        self.root = self.root.get_next()    #[1] check if first package
                        return True
                    else:
                        prev.set_next(current.get_next())   #[1] else: link prev to current.get_next()
                        return True
                else:
                    prev = current                          #[1] correct advancement
                    current = current.get_next()

    


deliver26may = Queue()
deliver26may.enqueue('AMZ001')
deliver26may.enqueue('LAZ001')
deliver26may.enqueue('AMZ002')
deliver26may.enqueue('SHP001')
deliver26may.enqueue('SHP002')
deliver26may.cancel('LAZ001')
deliver26may.display()
```

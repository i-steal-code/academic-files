```python
class Stack:
    def __init__(self):
        self.arr = []
        
    def push(self,value):
        self.arr.append(value)
        
    def pop(self):
        if self.arr:
            return self.arr.pop()
        else:
            print("No more moves")
            
    def display(self):
        if self.arr:
            print("Top of stack")
            index = 1
            while index <= len(self.arr):
                print(self.arr[-1*index])
                index += 1
        else:
            print("Stack is empty")
            
    def is_empty(self):
        return self.arr == []
            
# test stack
stack = Stack()
stack.display()

for n in range(3):
    stack.push(n)
stack.display()    

for n in range(2):
    print(stack.pop())
    
stack.push(5)
stack.display()

for n in range(3):
    stack.pop()
```

```python
from random import randint

class Knight:
    
    def __init__(self):
        self.X_pos = []
        self.K_pos = []
        while self.X_pos == self.K_pos:
            self.X_pos = [randint(1, 8) for n in range(2)]
            self.K_pos = [randint(1, 8) for n in range(2)]
        self.u_stack = Stack()
        self.grid()
    
    def grid(self):
        print(f'X position:{self.X_pos}, K position:{self.K_pos}')
        if self.X_pos == self.K_pos:
            print("Congratulations! You have won the game!")
        
        for y in range(9, 0, -1):
            for x in range(9):
                if y == 9:
                    if x == 0:
                        string = "  "
                    else:
                        string += f'{x} '
                else:
                    if x == 0:
                        string = f"{y}|"
                    elif x ==  self.X_pos[0] and y == self.X_pos[1]:
                        string += "X|"
                    elif x ==  self.K_pos[0] and y == self.K_pos[1]:
                        string += "K|"
                    else:
                        string += " |"
            print(string)
            print(" "+'-'*17)
            
    def move(self, k_col, k_row):
        if (abs(self.K_pos[0]-k_col) == 1 and abs(self.K_pos[1]-k_row) == 2) \
            or (abs(self.K_pos[0]-k_col) == 2 and abs(self.K_pos[1]-k_row) == 1):
            self.u_stack.push(self.K_pos)
            self.K_pos = [k_col, k_row]
        else:
            print("Invalid Move")
        self.grid()
        
    def undo(self):
        if not self.u_stack.is_empty():
            self.K_pos = self.u_stack.pop()
        else:
            print("No moves to undo")
        self.grid()
            
            
        
```

```python
game = Knight()
```

```python
game.move(4, 6)
```

```python
game.undo()
```

```python
game.undo()
```

```python
class Knight2(Knight):
    def move(self, k_col, k_row):
        if (abs(self.K_pos[0]-k_col) == 1 and abs(self.K_pos[1]-k_row) == 2) \
            or (abs(self.K_pos[0]-k_col) == 2 and abs(self.K_pos[1]-k_row) == 1):
            self.u_stack.push(self.K_pos)
            self.K_pos = [k_col, k_row]
            self.r_stack = Stack()
        else:
            print("Invalid Move")
        self.grid()    
    
    def undo(self):
        if not self.u_stack.is_empty():
            self.r_stack.push(self.K_pos)
            self.K_pos = self.u_stack.pop()
        else:
            print("No moves to undo")
        self.grid()
        
    def redo(self):
        if not self.r_stack.is_empty():
            self.u_stack.push(self.K_pos)
            self.K_pos = self.r_stack.pop()
        else:
            print("No moves to redo")
        self.grid()
            
```

```python
#Task 2.1
import sqlite3

# Create db connection
conn = sqlite3.connect("college.db")
cursor = conn.cursor()

# Create Student table
cursor.execute("""
CREATE TABLE IF NOT EXISTS Student (
    stu_id INTEGER PRIMARY KEY,
    name TEXT,
    civics_class TEXT,
    handphone TEXT);
""")

# Create Latecoming table
cursor.execute("""
CREATE TABLE IF NOT EXISTS Latecoming (
    date TEXT NOT NULL,
    stu_id INTEGER NOT NULL,
    reason TEXT,
    PRIMARY KEY (date, stu_id)
    FOREIGN KEY (stu_id) REFERENCES Student(stu_id));
""")

# Commit and close
conn.commit()
conn.close()

print("Tables created.")
```

```python
#Task 2.2
import csv

# Create db connection
conn = sqlite3.connect("college.db")
cursor = conn.cursor()

# Insert data into Student table
# File closes after with block
with open("STUDENT.csv", "r") as file:
    reader = csv.reader(file)
    next(reader)  # skip header
    for student in reader:
        cursor.execute("""
        INSERT INTO Student (stu_id, name, civics_class, handphone)
        VALUES (?, ?, ?, ?);
        """, student)

# Insert data into Latecoming table
# File closes after with block
with open("LATE.csv", "r") as file:
    reader = csv.reader(file)
    next(reader)  # skip header
    for late_record in reader:
        cursor.execute("""
        INSERT INTO Latecoming (date, stu_id, reason)
        VALUES (?, ?, ?);
        """, late_record)

# Commit and close
conn.commit()
conn.close()

print("Data has been inserted.")
```

```python
#Task 2.3 (Using GROUP BY, HAVING)
def filter_latecomers(civics_class, min_latecomings, start_date, end_date):
    conn = sqlite3.connect("college.db")
    cursor = conn.cursor()

    cursor.execute("""
    SELECT Student.name
    FROM Student
    INNER JOIN Latecoming ON Student.stu_id = Latecoming.stu_id
    WHERE Student.civics_class = ? AND Latecoming.date BETWEEN ? AND ?
    GROUP BY Student.name
    HAVING COUNT(*) >= ?
    ORDER BY Student.name;
    """, (civics_class, start_date, end_date, min_latecomings))
    
    students = cursor.fetchall()
    
    # Close connection
    conn.close()
    
    # Extract student names only
    student_names = [student[0] for student in students]
    
    return student_names

# Test
result = filter_latecomers("25S02Y", 2, "2024-08-01", "2024-08-31")
print(result)
```

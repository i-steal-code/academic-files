```python
#Task 3.1
def task3_1(quantity_of_data):
    if quantity_of_data.endswith("KB"):    #string ends with 'KB'
        x,y = quantity_of_data.split("KB") #x takes on the number 
        if x.isdigit():                    #x is a number
            return int(x)*1000             #multiply x by a thousand
        else:                              #x is not a number
            return "invalid data"          #error message
    elif quantity_of_data.endswith("MB"):
        x,y = quantity_of_data.split("MB")
        if x.isdigit():
            return int(x)*(1000000)
        else:
            return "invalid data"
    elif quantity_of_data.endswith("GB"):
        x,y = quantity_of_data.split("GB")
        if x.isdigit():
            return int(x)*1000000000
        else:
            return "invalid data"
    elif quantity_of_data.endswith("TB"):
        x,y = quantity_of_data.split("TB")
        if x.isdigit():
            return int(x)*1000000000000
        else:
            return "invalid data"
    else:
        return "invalid data"

print(task3_1("8KB"))
print(task3_1("8MB"))
print(task3_1("8GB"))
print(task3_1("8TB"))
```

```python
def task3_2(quantity_of_data):
    if quantity_of_data.endswith("KB"): #quantity_of_data is in KB
        x,y = quantity_of_data.split("KB")
        if x.isdigit():
            return int(x)*10**3
        else:
            return "invalid data"
    elif quantity_of_data.endswith("MB"): #quantity_of_data is in MB
        x,y = quantity_of_data.split("MB")
        if x.isdigit():
            return int(x)*(10**6)
        else:
            return "invalid data"
    elif quantity_of_data.endswith("GB"): #quantity_of_data is in GB
        x,y = quantity_of_data.split("GB")
        if x.isdigit():
            return int(x)*10**9
        else:
            return "invalid data"
    elif quantity_of_data.endswith("TB"): #quantity_of_data is in TB
        x,y = quantity_of_data.split("TB")
        if x.isdigit():
            return int(x)*10**12
        else:
            return "invalid data"
    elif quantity_of_data.endswith("KiB"): #quantity_of_data is in KiB
        x,y = quantity_of_data.split("KiB")
        if x.isdigit():
            return int(x)*2**10
    elif quantity_of_data.endswith("MiB"): #quantity_of_data is in MiB
        x,y = quantity_of_data.split("MiB")
        if x.isdigit():
            return int(x)*2**20
    elif quantity_of_data.endswith("GiB"): #quantity_of_data is in GiB
        x,y = quantity_of_data.split("GiB")
        if x.isdigit():
            return int(x)*2**30
    elif quantity_of_data.endswith("TiB"): #quantity_of_data is in TiB
        x,y = quantity_of_data.split("TiB")
        if x.isdigit():
            return int(x)*2**40
    else:                        #invalid format of quantity_of_data
        return "invalid data"    #error message

print(task3_2("512MiB")) #test function with '512MiB'
print(task3_2("2MiB"))
```

```python
def task3_3(quantity_of_data, target_unit):
    approved_units = ["KB","KiB","MB","MiB","GB","GiB","TB","TiB"]
    if target_unit not in approved_units: #invalid format of target_unit
        return "invalid data"
    actual_bytes = task3_2(quantity_of_data)
    if actual_bytes != "invalid data":
        if target_unit == 'KB':
            answer = int(actual_bytes)/1000
        elif target_unit == 'MB':
            answer = int(actual_bytes)/1000000
        elif target_unit == 'GB':
            answer = int(actual_bytes)/1000000000
        elif target_unit == 'TB':
            answer = int(actual_bytes)/1000000000000
        elif target_unit == 'KiB':
            answer = int(actual_bytes)/(2**10)
        elif target_unit == 'MiB':
            answer = int(actual_bytes)/(2**20)
        elif target_unit == 'GiB':
            answer = int(actual_bytes)/(2**30)
        elif target_unit == 'TiB':
            answer = int(actual_bytes)/(2**40)
    return answer
```

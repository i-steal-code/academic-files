```python
#Task 2.1 <- remember to include this as stated in QP
def task2_1(filename, quantity, maximum): #function declaration
    import random             #import module/library
    file = open(filename,"w") #open file for writing
    for i in range(quantity): #interate number of times (in quantity)
        num = random.randint(0,maximum) #generate random number between 0 and maximum inclusive
        file.write(str(num)+"\n") #output to file
    file.close()  #close file
    #if you use "with open", do include a comment to close file at the end

filename = "randomnumbers_name_3009_1234.txt" #define filename
task2_1(filename, 1000, 5000) #call task2_1
```

```python
#Task 2.2  <- remember to include this as stated in QP
def Merge(Left,Right):  #function declaration
    Array = [0 for i in range(len(Left)+len(Right))] #sorted result
    i,j,k = 0,0,0  #initialise i,j and k to be 0
    while i < len(Left) and j < len(Right): #elements remain in Left and Right array
        if Left[i] <= Right[j]: #1st element in Left is less than 1st element in Right
            Array[k] = Left[i] #copy first element from Left list to Array
            i = i+1
        else: #1st element in Right is less than 1st element in Left
            Array[k] = Right[j] #copy first element from Right list to Array
            j = j+1
        k = k+1
    while i < len(Left):    #element(s) remain on Left list
        Array[k] = Left[i]  #copy element(s) from Left list to Array
        i = i+1
        k = k+1
    while j < len(Right):   #element(s) remain on Right list
        Array[k] = Right[j] #copy element(s) from Right list to Array
        j=j+1
        k = k+1
    return Array  #return Array to calling function

def Mergesort(Array): #function declartion
    if len(Array) == 1: #only 1 element left in Array
        return Array
    mid = len(Array)//2           #find middle element
    left=[0 for i in range(mid)]  #define left array with half the size of Array
    right=[0 for i in range(mid, len(Array))] #define right array with the other half the size of Array
    for i in range(mid): #copy first half of "Array" into "left" array
        left[i] = Array[i]
    for i in range(mid, len(Array)): #copy second half of "Array" into "right" array
        right[i-mid] = Array[i]
    left = Mergesort(left) #recursive call of Mergesort, store return array to left
    right = Mergesort(right) #recursive call of Mergesort, store return array to right
    return Merge(left,right) #return array after calling Merge function

def task2_2(list_of_integers): #parameter is a list of integers
    sorted_list = Mergesort(list_of_integers) #call Mergesort function
    return sorted_list #return list

task2_2([56,25,4,98,0,18,4,5,7,0]) == [0,0,4,4,5,7,18,25,56,98] #testing function
        
```

```python
def task2_3(filename_in, filename_out): #function declaration, parameters are two files
    infile = open(filename_in,'r') #open file for reading
    num_list = []
    for line in infile: #iterate thru file line by line
        num_list.append(int(line.strip())) #add to the list num_list
    infile.close() #close file
    sorted_num_list = task2_2(num_list) #call function to sort num_list
    outfile = open(filename_out,'w') #open fie for writing
    for each in sorted_num_list:
        outfile.write(str(each)+'\n')
    outfile.close() #close file

task2_3("randomnumbers_name_3009_1234.txt","sortednumbers_name_3009_1234.txt")
```

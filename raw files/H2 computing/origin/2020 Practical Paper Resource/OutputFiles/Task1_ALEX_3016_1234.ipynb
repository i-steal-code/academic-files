```python
#Task 1.1  <- remember to include this as stated in QP
def task1_1(string_value): #function declaration, return integer
    h=0                                #initialise h value
    for i in range(len(string_value)): #iterate through string_value
        val = 33 * ord(string_value[i])
        h = (h+val) % 1024             #calculate hash value
    return h                           #return hash value
```

```python
task1_1("Hello") #testing task1_1
```

```python
task1_1("Hallo") #testing task1_1
```

```python
task1_1("Hullo") #testing task1_1
```

```python
#Task 1.2  <- remember to include this as stated in QP
def task1_2(seed, string_value): #function declaration, return integer
    join = seed + string_value  #concatenate two strings
    return task1_1(join)        #call task1_1 and return integer
```

```python
task1_2("seed-one","Hello")  #testing task1_2
```

```python
task1_2("seed-two","Hello")  #testing task1_2
```

```python
#Task 4.1

s = 100 #size of list
global hash_table #make the list a global object that can be called within a function
hash_table = [None for _ in range(s)] #populate table with null objects
```

```python
#Task 4.2

def calculate_hash(key):
    #i really have nothing to comment on this cell. i just copied the question paper's 'k mod s' requirement and thats all
    return int(key) % s
```

```python
#Task 4.3

def insert(key, data):
    index = calculate_hash(key)
    #check if list is full
    if None not in hash_table:
        return False 
    #increment index and probe linearly until we find free slot
    while index < s and hash_table[index] is not None:
        index += 1
        #handle full list case - we have walked off the list fully
        if index >= s:
            index = 0
    #insert into free slot
    hash_table[index] = [key,data]
    return True
```

```python
#Task 4.4

def read_data(filename):
    with open(filename) as file:
        for row in file:
            row = row.strip().split(' ') #clean each row in txt file
            insertion = insert(row[0],row[1])
            #insertion returned a false and it was not inserted properly
            if not insertion:
                print(f'item {row} was not inserted properly as the list is full')
            
read_data('records.txt')
print(hash_table)
```

```python
#Task 4.5

def find_record(key):
    index = calculate_hash(key)
    out = hash_table[index]
    #validate the record being returned
    if out is not None:
        return out
    else: #record is none and hence not stored in table
        return False
```

```python
#Task 4.6

read_data('records.txt')
place = find_record(475)
if place is False:
    print('Record not found')
else:
    print(f'key: {place[0]} | data: {place[1]}')
```

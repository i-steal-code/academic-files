```python
#Task 2.1

class Room:
    #construct object and clean its initialised data
    def __init__(self, north_exit, east_exit, south_exit, west_exit):
        self.__north_exit = int(north_exit)
        self.__east_exit = int(east_exit)
        self.__south_exit = int(south_exit)
        self.__west_exit = int(west_exit)
    
    #output respective index for whichever gate chosen
    def get_exit(self, gate):
        if gate == 'N':
            return self.__north_exit
        elif gate == 'E':
            return self.__east_exit
        elif gate == 'S':
            return self.__south_exit
        elif gate == 'W':
            return self.__west_exit
        else:
            return None
        
class Start(Room):
    #parent constructor with start text 
    def __init__(self, north_exit, east_exit, south_exit, west_exit, start_text):
        super().__init__(north_exit, east_exit, south_exit, west_exit)
        self.__start_text = str(start_text)
    
    #add a new message for every direction available
    def get_message(self):
        out = ['Start', self.__start_text]
        doors = ['N', 'E', 'S', 'W']
        for d in doors:
            if self.get_exit(d) != -1:
                out.append(f'there is door in direction {d}')
        return out
    
class Finish(Room):
    #parent constructor with finish text
    def __init__(self, north_exit, east_exit, south_exit, west_exit, finish_text):
        super().__init__(north_exit, east_exit, south_exit, west_exit)
        self.__finish_text = str(finish_text)
    
    #simple return of finish text and 'finish' header
    def get_message(self):
        return ['Finish', self.__finish_text]

class Normal(Room):
    #parent constructor with question and answer and clean init data
    def __init__(self, north_exit, east_exit, south_exit, west_exit, question, answer):
        super().__init__(north_exit, east_exit, south_exit, west_exit)
        self.__question = str(question)
        self.__answer = str(answer).lower() 
    
    #same way of doing things as start.get_message, but instead of appending another line we just add
    def get_message(self):
        out = ['Normal','']
        doors = ['N', 'E', 'S', 'W']
        for d in doors:
            if self.get_exit(d) != -1:
                out[1] += f'there is room to direction {d}. '
        return out
    
    #getter for question
    def get_question(self):
        return self.__question
    
    #compare input answer to object's stored answer
    def check_answer(self, uinpt):
        uinpt = uinpt.strip().lower() #clean user input 
        if uinpt == self.__answer:
            return True
        else:
            return False
```

```python
#Task 2.2

def seed_rooms(filename):
    with open(filename) as file: #read file
        rooms = [] #initialise list
        for row in file:
            row = row.strip().split(',') #clean row in txt file
            #insert an object into list according to their headers in the row
            if row[0] == 'Start':
                new = Start(row[1],row[2],row[3],row[4],row[5]) 
            elif row[0] == 'Finish':
                new = Finish(row[1],row[2],row[3],row[4],row[5])
            else:
                new = Normal(row[1],row[2],row[3],row[4],row[5],row[6])
            rooms.append(new)
        return rooms #return list
```

```python
#Task 2.3

def play():
    rooms = seed_rooms('rooms.txt')
    #find start and print start message (just in case someone decides to seed start not as the first file row)
    for n in rooms:
        if type(n) is Start:
            newroom = n
            message = n.get_message()[1:]
            for item in message:
                print(item)
            break
    doors = ['N', 'E', 'S', 'W']
    finished = False
    while not finished:
        #get the message and print it out
        out = newroom.get_message()
        if out[0] in ['Finish', 'Normal']:
            print(out[1])
        #handle finish
        if out[0] == 'Finish':
            finished = True
            return
        #user input to go through door until valid response
        valid = False
        while not valid:
            uinpt = input('please key in which door to go through: ')
            if uinpt in doors:
                nxt = newroom.get_exit(uinpt)
                #catch strays trying to go to non-existent doors
                if nxt == -1:
                    print('you cant go there!')
                else:
                    valid = True
            else:
                print('invalid input!')
        #if not finish, then it is a normal room
        if out[0] == 'Normal':
            print(newroom.get_question())
            #loop until correct answer is given
            correct = False
            while not correct:
                ans = input('your answer: ')
                correct = newroom.check_answer(ans)
            #answer is correct and we can let the user go
        #traverse to new room
        newroom = rooms[nxt]
    
    
play()
```

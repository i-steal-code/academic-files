```python
#Task 3.1

import sqlite3

#initialise everything
db = open('database.db','w')
db.close()
conn = sqlite3.connect('database.db')
cur = conn.cursor()

#reset tables
cur.execute("""
DROP TABLE IF EXISTS User
""")
cur.execute("""
DROP TABLE IF EXISTS Post
""")
cur.execute("""
DROP TABLE IF EXISTS Comment
""")

#define schema
cur.execute("""
CREATE TABLE User(
Username TEXT PRIMARY KEY,
Blocked INTEGER
)
""")
cur.execute("""
CREATE TABLE Post(
PostID INTEGER PRIMARY KEY,
PostText TEXT,
DateTimePosted TEXT,
Username TEXT,
FOREIGN KEY(Username) REFERENCES User
)
""")
cur.execute("""
CREATE TABLE Comment(
CommentID INTEGER PRIMARY KEY, 
PostID INTEGER,
Username TEXT, 
TheText TEXT,
DateTimeCommented TEXT,
FOREIGN KEY(PostID) REFERENCES Post,
FOREIGN KEY(Username) REFERENCES User
)
""")
```

```python
#Task 3.2

#seed data
with open('comment.txt') as file:
    for row in file:
        row = row.strip().split(',') #format each row in txt file properly
        cur.execute("""
        INSERT OR IGNORE INTO Comment(CommentID, PostID, Username, TheText, DateTimeCommented)
        VALUES (?,?,?,?,?)
        """, (row[0],row[1],row[2],row[3],row[4])) #sql query parameterisation 
        
with open('post.txt') as file: #do the same for the other files
    for row in file:
        row = row.strip().split(',')
        cur.execute("""
        INSERT OR IGNORE INTO Post(PostID, PostText, DateTimePosted, Username)
        VALUES (?,?,?,?)
        """, (row[0],row[1],row[2],row[3]))
        
with open('user.txt') as file:
    for row in file:
        row = row.strip().split(',')
        cur.execute("""
        INSERT OR IGNORE INTO User(Username, Blocked)
        VALUES (?,?)
        """, (row[0],row[1]))

conn.commit()
```

```python
#Task 1.1

#top of stack is treated as end of list
#top of stack is pointing at index that is at the end and is free

def push(stack, top, data):
    #increase bound of stack if top exceeds current size
    if top >= len(stack)-1:
        stack.append(None)
    #insert data as top
    stack[top] = data
    #update pointer
    top += 1
    return top
```

```python
#Task 1.2

def pop(stack, top):
    #check for empty stack
    if top == 0:
        return False
    elif stack[top-1] is None:
        return False
    #get the top 
    out = stack[top-1]
    #update pointer
    top -= 1
    return [out,top] #return data and pointer
```

```python
#Task 1.3

def main(number):
    #0 case
    if number == 0:
        print(0)
        return
    stack = []
    top = 0
    #regular case
    #push everything
    while number != 0:
        remain = number % 2
        number = number // 2
        top = push(stack,top,remain)
    #pop everything
    collector = ''
    add = True
    while add is not False:
        add = pop(stack,top)
        if add is not False:
            collector += str(add[0])
            top = add[1]
    print(collector)
    return

main(0)
main(22)
main(46967)
```

```python
#bubble sort
test = [1,3,4,6,2,5,5,7,3,6,3]

def bubblesort(arr):
    length = len(arr)
    swapped = True
    while swapped:
        swapped = False
        for n in range(length-1):
            if arr[n] > arr[n+1]:
                arr[n], arr[n+1] = arr[n+1], arr[n]
                swapped = True
    return arr

bubblesort(test)
```

```python
#insertion sort
test = [1,3,4,6,2,5,5,7,3,6,3]

def insertionsort(arr):
    for n in range(len(arr)):
        current = arr[n]
        j = n-1
        while j >= 0 and current < arr[j]:
            arr[j+1] = arr[j]
            j -= 1
        arr[j+1] = current
    return arr

insertionsort(test)
```

```python
#mergesort

def merge(arr,low,high):
    mid = (low + high) // 2
    left = arr[low : mid + 1]
    right  = arr[mid+1 : high+1]
    l = 0 
    r = 0
    a = low
    while l < len(left) and r < len(right):
        if left[l] < right[r]:
            arr[a] = left[l]
            l += 1
            a += 1
        else:
            arr[a] = right[r]
            r += 1
            a += 1
    while r < len(right):
        arr[a] = right[r]
        r += 1
        a += 1
    while l < len(left):
        arr[a] = left[l]
        l += 1
        a += 1
    return arr

def mergesort(arr,low,high):
    if low < high:
        mid = (low + high) // 2
        mergesort(arr,low,mid)
        mergesort(arr,mid+1,high)
        merge(arr,low,high)
    return arr

test = [1,3,4,6,2,5,5,7,3,6,3]
mergesort(test, 0, len(test)-1)
```

```python
#quicksort

def partition(arr,first,last):
    pivot = arr[first]
    left = first + 1
    right = last
    done = False
    while not done:
        while left <= right and arr[left] <= pivot:
            left += 1
        while right >= left and arr[right] >= pivot:
            right -= 1
        if left > right:
            done = True
        else:
            arr[left],arr[right] = arr[right],arr[left]
    arr[right],arr[first] = pivot,arr[right]
    return right

def quicksort(arr,first,last):
    if first < last:
        split = partition(arr,first,last)
        quicksort(arr,first,split-1)
        quicksort(arr,split+1,last)
    return arr

test = [1,3,4,6,2,5,5,7,3,6,3]
quicksort(test,0,len(test)-1)
```

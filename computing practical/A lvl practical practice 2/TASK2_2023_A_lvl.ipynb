```python
#Task 2.1

#declare node object

class QueueItem:
    def __init__(self, ID = 'Null', quantity = 'Null'):
        self.__ID = ID
        self.__quantity = quantity
        
    def getID(self):
        return self.__ID
    
    def getQuantity(self):
        return self.__quantity
```

```python
#Task 2.2

#declare circular queue (static type)

class Queue:
    def __init__(self, numberItems = 4):
        self.__queueList = [None for _ in range(numberItems)]
        self.__headPointer = 0
        self.__tailPointer = 0 #since queue size is 4, largest index is 3
        self.__numberItems = numberItems
        
    def enqueue(self, item):
        #validate whether there is still space or not
        if None not in self.__queueList:
#             print('we ran out of buffer')
            return False
        #there is a free slot and we can therefore insert at tail
        self.__queueList[self.__tailPointer] = item
        #increment tail pointer
        self.__tailPointer += 1
        #circle back
        if self.__tailPointer > 3:
            self.__tailPointer = 0
        return True
    
    def dequeue(self):
        #pop queue head
        output = self.__queueList[self.__headPointer]
        self.__queueList[self.__headPointer] = None
        #handle empty list
        if self.__queueList is [None for _ in range(self.__numberItems)]:
#             print('this thing is empty')
            return False
        #increment head pointer
        self.__headPointer += 1
        #circle back
        if self.__headPointer > 3:
            self.__headPointer = 0
        return output
    
    def printQueue(self):
        for i in self.__queueList:
            try:
                print(f'{i.getID()} | {i.getQuantity()}')
            except:
                print('none')
        
```

```python
#Task 2.3

#test enqueue

test = Queue()
for i in range(5):
    ID = input('key in id: ')
    quant = input('key in quantity: ')
    result = test.enqueue(QueueItem(ID,quant))
    if result:
        print('successful enqueue')
    else:
        print('unsuccessful enqueue')
```

```python
#Task 2.4

#test dequeue

for i in range(5):
    result = test.dequeue()
    if result:
        print(result.getID(), result.getQuantity())
    else:
        print('dequeue cannot be performed as queue is empty')
```

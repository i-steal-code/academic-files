```python
#Task 2.1
import pymongo, json
client = pymongo.MongoClient("mongodb://localhost:27017")
attendance = client["college_db"]["attendance"]
attendance.delete_many({})
with open("attendance.json") as file:
    data = json.load(file)
    for r in data:
        attendance.insert_one(r)
print("successfully imported data into database")
```

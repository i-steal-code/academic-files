```python
#Task 1.1
import sqlite3, csv
conn = sqlite3.connect('college.db')
cur = conn.cursor()

cur.execute("""
DROP TABLE IF EXISTS Student
""")
cur.execute("""
DROP TABLE IF EXISTS Late
""")

cur.execute("""
CREATE TABLE IF NOT EXISTS Student(
stu_id INTEGER PRIMARY KEY,
name TEXT,
civics_class TEXT,
handphone INTEGER
)
""")
cur.execute("""
CREATE TABLE IF NOT EXISTS Late(
date DATE,
stu_id PRIMARY KEY,
reason TEXT,
FOREIGN KEY(stu_id) REFERENCES Student(stu_id)
)
""")
print("successfully created tables")
```

```python
#Task 1.2
with open("STUDENT.csv") as file:
    data = csv.reader(file)
    h = next(data)
    for r in data:
        cur.execute(f"""
        INSERT OR IGNORE INTO Student
        (stu_id,name,civics_class,handphone)
        VALUES (?,?,?,?)
        """, (r[0],r[1],r[2],r[3]))
        
with open("LATE.csv") as file:
    data = csv.reader(file)
    h = next(data)
    for r in data:
        cur.execute(f"""
        INSERT OR IGNORE INTO Late
        ({h[0]},stu_id,{h[2]})
        VALUES (?,?,?)
        """, (r[0],r[1],r[2]))

conn.commit()
print("successfully inserted data into tables")
```

```python
#Task 1.3
def filter_latecomers(civics_class,min_latecomings, start_date, end_date):
    cur.execute("""
    SELECT Student.name
    FROM Late LEFT OUTER JOIN Student ON Student.stu_id = Late.stu_id
    WHERE Student.civics_class = ? 
    AND Late.date <= ? 
    AND Late.date >= ?
    GROUP BY Late.date
    HAVING COUNT(Late.stu_id) >= ?
    ORDER BY Student.name ASC;
    """, (civics_class, end_date, start_date, min_latecomings))
    data = cur.fetchall()
    output = []
    for i in data:
        output.append(i[0])
    return output

print(filter_latecomers("25S02Y",2,"2024-08-01","2024-08-31"))
```

## Part B1 — OOP hierarchy (12 min)

Implement:

- `Resource(resource_id, title)` with private `__resource_id`, `__title`, getters, `set_title`, and `describe()` → `Resource {id}: {title}`
- `DigitalResource(resource_id, title, filesize_mb)` inheriting from `Resource`: call `super().__init__`, store private `__filesize_mb` as `int`, override `describe()` using **getters** for parent fields → `Digital Resource {id}: {title} ({n} MB)`

Build `stock`: a Python list of at least 6 mixed Resource / DigitalResource instances. Print each `.describe()`.


```python
# Part B1 — YOUR ATTEMPT

stock = []

class Resource:
    def __init__(self, resource_id, title):
        self.__resource_id = resource_id
        self.__title = title
        
    def set_title(self, title):
        self.__title = title
        
    def get_title(self):
        return self.__title
    
    def get_id(self):
        return self.__resource_id
    
    def describe(self):
        return f"Resource {self.__resource_id}: {self.__title}"
    
class DigitalResource(Resource):
    def __init__(self, resource_id, title, filesize_mb):
        super().__init__(resource_id, title)
        self.__filesize_mb = int(filesize_mb)
        
    def get_size(self):
        return self.__filesize_mb
    
    def set_size(self, size):
        self.__filesize = size
        
    def describe(self):
        return f"DigitalResource {super().get_id()}: {super().get_title()} ({self.__filesize_mb} MB)"
    
i = 0
while i <= 12:
    stock.append(Resource(i, "resource"))
    stock.append(DigitalResource(i+1, "digital", 67))
    print(stock[i].describe())
    print(stock[i+1].describe())
    i += 2
    
```

## Part B2 — LinkedList of resource IDs (15 min)

`Node` (data + right link, getters/setters) and `LinkedList` with `__start`, `is_empty`, `insert_end`, `search` (Node or None), `delete` (**including head case**).

Insert every `resource_id` from `stock` via `insert_end` into `catalogue = LinkedList()`.

Then: `delete` the **first** id you inserted (head), `search` a mid id (found), `search(-1)` → None, `delete` a missing id (no crash).


```python
# Part B2 — YOUR ATTEMPT

catalogue = None  # LinkedList()

class Node:
    def __init__(self, data):
        self.__right = None
        self.__data = data
        
    def get_data(self):
        return self.__data
    
    def set_data(self, data):
        self.__data = data
        
    def get_right(self):
        return self.__right
    
    def set_right(self, right):
        self.__right = right
        
class LinkedList:
    def __init__(self, start = None):
        self.__start = start
        
    def is_empty(self):
        return bool(self.__start)
    
    def insert_end(self, new_data):
        new = Node(new_data)
        if self.__start is None:
            self.__start = new
            return
        else:
            current = self.__start
            while current.get_right(): #traverse until current is last node
                current = current.get_right()
            current.set_right(new)
            return
        
    def search(self, data):
        current = self.__start
        while current and current.get_data().get_id() != data: #if current exists, loop linear search
            current = current.get_right()
        return current 
    
    def delete(self, data):
        if self.__start is not None and self.__start.get_right() is not None:
            current = self.__start
            while current and current.get_data().get_id() != data: #if current exists, loop linear search
                previous = current
                current = current.get_right()
            if current:
                previous.set_right(current.get_right()) #logic assumes there is a previous and current, and current is found
            else: #current is none; reached end of list without finding data
                return None 
        else:
            self.__start = None
    
catalogue = LinkedList()

for i in stock:
    catalogue.insert_end(i)

catalogue.delete(1)
catalogue.search(10)
catalogue.search(-1)
catalogue.delete(-2)
```

## Part B4 — sort + binary search on extracted keys (12 min)

Build `keys` (list of ints) from the catalogue.

1. **Insertion sort** *or* `partition` + `quickSort` (final pivot↔rightmark swap). Sort a **copy** into `sorted_keys`.
2. Recursive `BinarySearch(Arr, FindValue, Low, High)` — `Low > High` → `-1`; recurse with `middle ± 1` and **return** the call.

Demo: search existing id and missing id on `sorted_keys`.


```python
# Part B4 — YOUR ATTEMPT

keys = [1,4,2,53,7,3,4342,3,3335,6,4]
sorted_keys = []

def partition(arr, first, last):
    pivotvalue = arr[first]
    leftmark = first + 1
    rightmark = last
    done = False
    while not done:
        while leftmark <= rightmark and arr[leftmark] <= pivotvalue:
            leftmark += 1
        while rightmark >= leftmark and arr[rightmark] >= pivotvalue:
            rightmark -= 1
        if leftmark > rightmark: #done when bounds cross
            done = True
        else:
            #swap objects at leftmark and rightmark if bounds have not crossed    
            temp = arr[leftmark]
            arr[leftmark] = arr[rightmark]
            arr[rightmark] = temp  
    #swap pivotvalue and rightmark object to put pivotvalue at its final sorted position
    arr[first] = arr[rightmark]
    arr[rightmark] = pivotvalue
    return rightmark
        
def quickSort(array, first, last): #divide and conquer
    if first < last:
        split = partition(array, first, last)
        quickSort(array, first, split - 1) 
        quickSort(array, split + 1, last)
    return array
    
sorted_keys = quickSort(keys, 0, len(keys)-1)

def BinarySearch(arr, findvalue, low, high):
    if low > high:
        return -1
    middle = (low + high) // 2
    if arr[middle] == findvalue: #we have a hit!
        return middle
    elif findvalue < arr[middle]: #go left
        return BinarySearch(arr, findvalue, low, middle - 1)
    else: #go right
        return BinarySearch(arr, findvalue, middle + 1, high)

BinarySearch(sorted_keys, 6, 0, len(sorted_keys)-1)
```

## Part B5 — JSON round-trip (8 min)

Build a JSON-serialisable list of dicts from `stock`, e.g.
`{"id": ..., "title": ..., "kind": "print"|"digital", "filesize_mb": ...|null}`.

- `save_catalogue(filename, data)` — `json.dump`, utf-8
- `load_catalogue(filename)` — `json.load`

Write `library_out.json`, load it back, assert same length.

When finished practising, delete `library_out.json` so it does not linger as a trail.


```python
# Part B5 — YOUR ATTEMPT
import json

def save_catalogue(filename, data):
    with open(filename,'w', encoding='utf-8') as f:
        insertion = []
        for row in stock:
            if type(row) is DigitalResource:
                kind = 'digital'
                filesize = row.get_size()
            else:
                kind = 'print'
                filesize = None
            insertion.append({"id": row.get_id(), "title": row.get_title(), "kind": kind, "filesize_mb": filesize})
        json.dump(insertion, f)
    
def load_catalogue(filename):
    with open(filename, encoding='utf-8') as f:
        return json.load(f)
        
save_catalogue("library_out.json", stock)
load_catalogue("library_out.json")
```

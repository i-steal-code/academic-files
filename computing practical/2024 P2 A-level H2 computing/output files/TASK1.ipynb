```python
#Task 1.1

import random

colour = ['red', 'green', 'blue']
p1 = []
p2 = []

for i in range(10):
    p1.append([random.randint(1,4),colour[random.randint(0,2)]])
    p2.append([random.randint(1,4),colour[random.randint(0,2)]])
    
print(p1,'\n',p2)
```

```python
#Task 1.2

def compare(first,second):
    first = colour.index(first)
    second = colour.index(second)
    #same
    if first == second:
        return '3'
    #blue first beat red second
    elif first == 2 and second == 0:
        return '1'
    #blue second beat red first
    elif second == 2 and first == 0:
        return '2'
    #regular cases
    #second smaller than first and therefore win
    elif second < first:
        return '2'
    #first smaller than secon and wins
    elif first < second:
        return '1'
    
```

```python
#Task 1.3

#bubble sort
def bubblesort(arr):
    length = len(arr)
    swapped = True
    while swapped:
        swapped = False
        for n in range(length-1):
            #compare card value first
            if arr[n][0] > arr[n+1][0]:
                temp = arr[n]
                arr[n] = arr[n+1]
                arr[n+1] = temp
                swapped = True
            #card values are the same
            elif arr[n][0] == arr[n+1][0]:
                #compare card colour
                if colour.index(arr[n][1]) > colour.index(arr[n][1]):
                    temp = arr[n]
                    arr[n] = arr[n+1]
                    arr[n+1] = temp
                    swapped = True
    return arr
        
print(p1)
print(bubblesort(p1))
```

```python
#Task 1.4

p1 = bubblesort(p1)
p2 = bubblesort(p2)

def calculate_score(p1,p2,cardcount):
    score1 = 0
    score2 = 0
    for n in range(cardcount):
        c = compare(p1[n][1],p2[n][1])
        if c == '1': #p1 win
            score1 += p1[n][0]
        elif c == '2': #p2 win
            score2 += p2[n][0]
        score1 += p1[n][0]
        score2 += p2[n][0]
    return (score1,score2)
        
result = calculate_score(p1,p2,10)
print(f'p1 hand: {p1} | p1 score: {result[0]}')
print(f'p2 hand: {p2} | p2 score: {result[1]}')
```

```python
#Task 1.5

def winner(score1, score2):
    if score1 == score2:
        print('tie')
    elif score1 > score2:
        print('p1 wins')
    else:
        print('p2 wins')

winner(result[0], result[1])
```

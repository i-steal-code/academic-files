```python
#Task 2.1

class Node:
    def __init__(self, data = -1, left = -1, right = -1):
        self.__data_value = data
        self.__left_node = left
        self.__right_node = right
        
    def get_data(self):
        return self.__data_value
    
    def set_data(self,data):
        self.__data_value = data
        
    def get_left(self):
        return self.__left_node
    
    def set_left(self,left):
        self.__left_node = left
        
    def get_right(self):
        return self.__right_node
    
    def set_right(self,right):
        self.__right_node = right
        
```

```python
#Task 2.2

def calculate_nodes(levels):
    nodes = 0
    for i in range(levels):
        nodes += 2**(i)
    print(f'A {levels} level number tree has {nodes} nodes')
    return nodes
```

```python
#Task 2.3

calculate_nodes(1)
calculate_nodes(3)
calculate_nodes(5)
calculate_nodes(10)
```

```python
#Task 2.4

def create_empty_tree(nodes,levels):
    nodes = [Node() for _ in range(calculate_nodes(levels))]
    return nodes
```

```python
#Task 2.5

def assign_leaf_nodes(nodes, levels):
    #find out how many leaves in bottom layer
    bottom = 2 ** levels-1
    #seed leaf nodes
    for n in range(bottom):
        nodes[n].set_data(n+1)
    return nodes
```

```python
#Task 2.6

def create_number_tree(nodes, levels):
    #since we know layer {levels} has already been populated, we will be working with everything above it
    total = calculate_nodes(levels)
    current = 2 ** (levels - 1) #we start current after the bottom layer leaves
    for n in range(total//2):
        temp = 0
        #add left and right leaf data
        temp += nodes[2*n].get_data()
        temp += nodes[2*n+1].get_data()
        #set current node to the added total
        nodes[current].set_data(temp)
        #set current node left and right to leaf nodes
        nodes[current].set_left(2*n)
        nodes[current].set_right(2*n+1)
        current += 1
        #move up current 
    return nodes
```

```python
#Task 2.7

#level 1
calculate_nodes(1)
test = create_empty_tree([],1)
test = assign_leaf_nodes(test,1)
test = create_number_tree(test,1)
for item in test:
    print(f'{item.get_left()} | {item.get_data()} | {item.get_right()}')
```

```python
#level 3
calculate_nodes(3)
test = create_empty_tree([],3)
test = assign_leaf_nodes(test,3)
test = create_number_tree(test,3)
for item in test:
    print(f'{item.get_left()} | {item.get_data()} | {item.get_right()}')
```

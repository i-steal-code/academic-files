```python
#Task 3.1

def heading(text):
    return f'<h1>{text}</h1>'

def subheading(text):
    return f'<h2>{text}</h2>'

def paragraph(text):
    return f'<p>{text}</p>'

print(heading('My element'))
print(subheading('My element'))
print(paragraph('My element'))
```

```python
#Task 3.2

def hyperlink(text):
    text = text.strip().split(',')
    return f'<a href="{text[0]}">{text[1]}</a>'

print(hyperlink('https://www.cambridgeinternational.org,Click here'))
```

```python
#Task 3.3

def body(convert):
    if convert[0] == 'heading':
        return heading(convert[1])
    elif convert[0] == 'subheading':
        return subheading(convert[1])
    elif convert[0] == 'paragraph':
        return paragraph(convert[1])
    elif convert[0] == 'hyperlink':
        return hyperlink(convert[1])
    else:
        #unrecognised element
        return False
```

```python
#Task 3.4

def read_data(filename):
    with open(filename) as file:
        out = ''
        for row in file:
            row = row.strip().lower().split(': ')
            out += body(row)
        return f'<html>{out}</html>'

def write_data(filename,converted):
    try:
        output = open(filename + '.html','w')
        output.write('<!doctype html>' + converted)
    except Exception as e:
        output = False
        print(f'unable to convert: {e}')
    return output

write_data('test',read_data('firstPage.txt'))
write_data('theWebsite',read_data('website.txt'))
```

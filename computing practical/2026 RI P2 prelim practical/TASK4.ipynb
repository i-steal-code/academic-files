```python
#Task 4.1

#declare node object
class Presentation:
    def __init__(self, OrderNumber, TeamName, ProjectTitle, Category):
        self.__OrderNumber = OrderNumber
        self.__TeamName = TeamName
        self.__ProjectTitle = ProjectTitle
        self.__Category = Category
        self.__NextPresentation = None
        
    def getOrderNumber(self):
        return self.__OrderNumber
    
    def getTeamName(self):
        return self.__TeamName
    
    def getProjectTitle(self):
        return self.__ProjectTitle
    
    def getCategory(self):
        return self.__Category
    
    def getNext(self):
        return self.__NextPresentation
    
    def setNext(self, p):
        self.__NextPresentation = p
        
    def display(self):
        prt = [self.__OrderNumber, self.__TeamName, self.__ProjectTitle, self.__Category]
        return prt
        
test = Presentation(2,3,5,2)
test.display()
```

```python
#Task 4.2 (in construction)

#declare linked list class
class PresentationSchedule:
    def __init__(self):
        self.__FirstPresentation = None
        
    def getFirst(self):
        return self.__FirstPresentation
    
    def setFirst(self, p):
        self.__FirstPresentation = p
        
    def AddPresentation(self, NewPresentation, current = self.getFirst()):
        
        #NewPresentation is the new node object being inserted
        nxt = current.getNext()
        #resolve edge cases first
        #reached the end; catches Null node that would otherwise break edge cases
        if nxt is None:
            #insert at endpoint
            current.setNext(NewPresentation)
            return True
        #duplicate OrderNumber
        elif current.getOrderNumber() == NewPresentation.GetOrderNumber():
            print("Duplicate OrderNumber")
            return False
        #completion cases
        #correct OrderNumber found, insert into current node's pointer and set nxt as next pointer
        elif NewPresentation.GetOrderNumber() < nxt.GetOrderNumber():
            """
            insert because we know from the previous pass that NewPresentation is greater than Current in order number
            we also know that this the new node is not equal to the current number because there is literally an edgeguard above
            we can conclude that in this situation, the NewPresentation's order number is definitely not a duplicate, 
            is greater than current order number, and is less than the next order number
            therefore, this is the completion case where we insert
            """ 
            current.setNext(NewPresentation)
            NewPresentation.setNext(nxt)
            return True
        else: #if all conditions do not pass, recurse
            return self.AddPresentation(NewPresentation, nxt) #traverse by recursing with nxt node as the new current node
    
    def summary(self):
        current = self.getFirst()
        size = 0
        categorydata = {}
        #traverse the list, display and count size. add all categories into anther list for further processing
        while current:
            data = current.display()
            size += 1
            if current.getCategory() in categorydata.keys():
                # +1 count to category if it exists in category data
                categorydata[current.getCategory()] += 1
            else: 
                #create new category and init as 1 in the dictionary
                categorydata.update({current.getCategory(): 1})
            current = current.getNext()
        #print size and category data
        print(f"number of presentations is {size}")
        for row in categorydata:
            print(row)
        return 
```

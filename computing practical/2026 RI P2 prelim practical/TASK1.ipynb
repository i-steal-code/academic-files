```python
#Task 1.1

def LoadWarehouse(Filename):
    warehouse = []
    i = 0
    with open(Filename, 'r') as file:
        for row in file:
            curr = row.split()
            warehouse.append(curr)
            if 'R' in row:
                robotcol = row.index('R') - 1
                robotrow = i
            else:
                i += 1
        file.close()
    return warehouse, robotrow, robotcol
        
LoadWarehouse("WAREHOUSE2.txt")
```

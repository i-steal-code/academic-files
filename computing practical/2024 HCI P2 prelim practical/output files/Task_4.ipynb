```python
#Task 4.1

#init
import sqlite3

conn = sqlite3.connect('TRIP.db')
conn.row_factory = sqlite3.Row
cur = conn.cursor()

cur.execute("""
DROP TABLE IF EXISTS Flight
""")
cur.execute("""
DROP TABLE IF EXISTS Customer
""")
cur.execute("""
DROP TABLE IF EXISTS Ticket
""")

#schema
cur.execute("""
CREATE TABLE Flight(
FlightNo TEXT PRIMARY KEY,
DepartCity TEXT,
ArrivalCity TEXT,
DepartTime TEXT,
ArrivalTime TEXT
)
""")
cur.execute("""
CREATE TABLE Customer(
CustomerNo INTEGER PRIMARY KEY,
Name TEXT,
Gender TEXT,
DOB INTEGER
)
""")
cur.execute("""
CREATE TABLE Ticket(
TicketNo INTEGER PRIMARY KEY,
CustomerNo INTEGER,
FlightNo TEXT,
Date INTEGER,
Seat TEXT,
FOREIGN KEY (CustomerNo) REFERENCES Customer,
FOREIGN KEY (FlightNo) REFERENCES Flight
)
""")
```

```python
#Task 4.2

#seed
with open('FLIGHT.txt') as file:
    for row in file:
        row = row.strip().split(',')
        cur.execute("""
        INSERT OR IGNORE INTO Flight(FlightNo, DepartCity, ArrivalCity, DepartTime, ArrivalTime)
        VALUES(?,?,?,?,?)
        """, (row[0],row[1],row[2],row[3],row[4]))
        
with open('CUSTOMER.txt') as file:
    for row in file:
        row = row.strip().split(',')
        cur.execute("""
        INSERT OR IGNORE INTO Customer(CustomerNo, Name, Gender, DOB)
        VALUES(?,?,?,?)
        """, (row[0],row[1],row[2],row[3]))
        
with open('TICKET.txt') as file:
    for row in file:
        row = row.strip().split(',')
        cur.execute("""
        INSERT OR IGNORE INTO Ticket(TicketNo, CustomerNo, FlightNo, Date, Seat)
        VALUES(?,?,?,?,?)
        """, (row[0],row[1],row[2],row[3],row[4]))
        
conn.commit()
```

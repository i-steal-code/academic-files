```python
#task 3.1

class TaskNode:
    def __init__(self, taskName, priority, nxt = None):
        self.__taskName = taskName
        self.__priority = priority
        self.__next = nxt
        
    def get_taskName(self):
        return self.__taskName
    
    def set_taskName(self, taskName):
        self.__taskName = taskName
        
    def get_priority(self):
        return self.__priority
    
    def set_priority(self, priority):
        self.__priority = priority
        
    def get_next(self):
        return self.__next
    
    def set_next(self, nxt):
        self.__next = nxt
        
    def __str__(self):
        return f"{self.__taskName},{self.__priority}"
```

```python
#Task 3.2

class PriorityQueue:
    def __init__(self, head = None):
        self.__head = head
        
    def viewPendingTasks(self):
        current = self.__head
        while current:
            print(current.__str__())
            current = current.get_next()
    
    def enqueue(self, taskNode):
        current = self.__head
        #empty head case
        if current is None:
            self.__head = taskNode
        #
        #regular middle of list cases 
        while current:
            #insert in current's place when it is larger than current (FCFS basis)
            if taskNode.get_priority() > current.get_priority():
                taskNode.set_next(current.get_next())
                current.set_next(taskNode)
                return True
            #reached end
            if not current.get_next():
                current.set_next(taskNode)
            #if not, traverse  
            current = current.get_next()
        
    def dequeue(self):
        dq = self.__head
        if dq:
            self.__head = dq.get_next()
            return dq
        else:
            return None
        
    def isEmpty(self):
        return bool(self.__head)
    
    def delete(self, taskName):
        #empty head
        if self.__head is None:
            return False
        #head is target
        if self.__head.get_taskName() == taskName:
            self.__head = self.__head.get_next()
            return True
        current = self.__head.get_next()
        previous = self.__head
        while current:
            if current.get_taskName() == taskName:
                previous.set_next(current.get_next())
                return True
            previous = current
            current = current.get_next()
        return False
        
    def search(self, taskName):
        current = self.__head
        if current is None:
            return None
        while current:
            if current.get_taskName() == taskName:
                return current
            else:
                current = current.get_next()
        return None
        
    def updatePriority(self, taskName, newPriority):
        #delete the task and then re-enqueue the task with new priority
        self.delete(taskName)
        self.enqueue(TaskNode(taskName, newPriority))
        return 
            
```

```python
#seed tasks
with open('TASKS.txt') as file:
    tasks = PriorityQueue()
    for row in file:
        row = row.strip().split(',')
        new = TaskNode(row[0],row[1])
        tasks.enqueue(new)
    tasks.viewPendingTasks()
    tasks.dequeue()
    tasks.dequeue()
    print('\n')
    tasks.viewPendingTasks()
    
```

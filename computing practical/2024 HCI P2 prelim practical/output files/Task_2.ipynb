```python
#Task 2.1

import csv

def read_csv(filename):
    with open(filename) as f:
        file = csv.reader(f)
        output = []
        for row in file:
            output.append([int(row[0]),int(row[1]),int(row[2]),row[3]])
        return output
        
colours = read_csv("paint.csv")
print(len(colours))
print(colours)
```

```python
#Task 2.2

def find(array,r,g,b):
    for item in array:
        if item[0] == r and item[1] == g and item[2] == b:
            return item[3]
    return 'nil'

print(find(colours, 253, 253, 5))
print(find(colours, 5, 253, 253))
```

```python
#Task 2.3

def rgb(array, colour):
    #label index + validation check
    if colour is 'red':
        index = 0
    elif colour is 'green':
        index = 1
    elif colour is 'blue':
        index = 2
    else:
        print('invalid colour given')
        return []
    #bubble sort
    done = False
    while not done:
        done = True
        for n in range(len(array)):
            if array[n][index] < array[n + 1][index]:
                #swap when wrong pos
                temp = array[n]
                array[n] = array[n + 1]
                array[n + 1] = temp
                done = False
            
    
colours = read_csv("paint.csv")
print(rgb(colours, 'yellow'))
print(rgb(colours, 'blue'))
```

```python
#Task 2.4

def partition(arr, first, last):
    pivot = (arr[first][0] + arr[first][1] + arr[first][2])
    left = first + 1
    right = last
    done = False
    while not done:
        while left <= right and (arr[left][0] + arr[left][1] + arr[left][2]) >= pivot:
            left += 1
        while right >= left and (arr[right][0] + arr[right][1] + arr[right][2]) <= pivot:
            right -= 1
        if left > right:
            done = True
        else:
            temp = arr[right]
            arr[right] = arr[left]
            arr[left] = temp
    temp = arr[first]
    arr[first] = arr[right]
    arr[right] = temp
    return right

def quicksort(arr, first, last):
    if first < last:
        split = partition(arr, first, last)
        quicksort(arr, first, split - 1)
        quicksort(arr, split + 1, last)
    return arr

def bright(array):
    return quicksort(array,0,len(array)-1)

colours = read_csv("paint.csv")
print(bright(colours))
```

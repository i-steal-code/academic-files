```python
#Task 1.1

def task1_1(key, message):
    length = len(key)
    #fill in spaces with z 
    message = message.replace(' ', 'z')
    #fill empty slots with x
    left = len(message) % length
    message += 'x' * left
    #split into horizontal rows
    collector = ''
    fin = []
    for i in message:
        if len(collector) < length:
            collector += i
        elif len(collector) == length:
            fin.append(collector)
            collector = i
    #form into vertical columns
    vert = [n for n in key]
    for row in fin:
        n = 0
        for i in row:
            vert[n] += i
            n += 1
    #organise vertical columns 
    vert.sort()
    #concatenate output while removing key headers
    out = ''
    for column in vert:
        out += column[1:]
    return out
```

```python
#Task 1.2

task1_1('ALGORITHM', 'computing is an interesting subject')
```

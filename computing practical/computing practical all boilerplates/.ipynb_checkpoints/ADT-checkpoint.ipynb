### Q1 work — Node


```python
class Node:
    def __init__(self, data=None):
        self.__data = data
        self.__right = None
    
    def get_data(self):
        return self.__data
    def set_data(self, data):
        self.__data = data
        
    def get_right(self):
        return self.__right
    def set_right(self, right):
        self.__right = right
```

### Q1 work — LinkedList + driver


```python
class LinkedList:
    def __init__(self, start = None):
        self.__start = start #starting pointer which is starting node object
        
    def is_empty(self):
        if self.__start == None:
            return True
        else:
            return False
    
    def insert_end(self, data):
        new = Node()
        new.set_data(data)
        if self.is_empty():
            self.__start = new
        else:
            current = self.__start
            while current.get_right() != None:
                current = current.get_right()
            current.set_right(new)
    
    def insert_front(self, data):
        new = Node()
        new.set_data(data)
        if self.is_empty():
            self.__start = new
        else:
            new.set_right(self.__start)
            self.__start = new
    
    def search(self, data):
        #linear search by traversing linear graph 
        current = self.__start
        while current is not None and current.get_data() != data:
            current = current.get_right()
        return current 
    
    def delete(self, data):
        #handle edge cases
        if self.is_empty():
            return None
        elif self.__start.get_data() == data:
            self.__start = self.__start.get_right()
        #linear search by traversing linear graph 
        previous = None
        current = self.__start
        while current and current.get_data() != data:
            previous = current
            current = current.get_right()
        if previous:
            previous.set_right(current.get_right())
    
    def display(self):
        prtlist = ''
        current = self.__start
        while current:
            prtlist += f' -> {current.get_data()}'
            current = current.get_right()
        return prtlist
        
```

```python
test = LinkedList()
for i in [50, 20, 70, 10, 40]:
    test.insert_end(i) #testing both front and end variants work as intended
test.display()
```

### Q1 work — TreeNode, BST + driver


```python
class TreeNode(Node):
    def __init__(self, data=None):
        super().__init__(data)
        self.__left = None
        
    def get_left(self):
        return self.__left
    def set_left(self, left):
        self.__left = left
    
class BST:
    def __init__(self, root=None):
        self.__root = root #treenode object
    
    #parents are any node that has at least 1 child. this means that one child could be a node but another would be empty.
    #children are nodes/phantom nodes that only have a parent and no child
    #parents are children that have children. this means they have a parent and a child themselves.
    #only the root does not have a parent
            
    
    #data is data within node
    def insert(self, data):
        new = TreeNode(data)
        if self.__root is None:
            self.__root = new
        else:
            #rule change to simplify code; all non-null nodes are parents since they are expecting in the insertion situation 
            child = self.__root
            #traverse binary tree by turning children into parents until end child is empty 
            while child:
                #child recognised as parent of next child node because it has at least 1 child 
                parent = child
                if data <= child.get_data():
                    #data seniority 
                    child = child.get_left()
                    #insert if child is none
                    if child == None:
                        parent.set_left(new)
                        return
                elif data > child.get_data():
                    child = child.get_right()
                    #insertion for right
                    if child == None:
                        parent.set_right(new)
                        return
                    
            
    def search(data):
        current = self.__root
        while current is not None:
            if data == current.get_data():
                return current
            elif data < current.get_data():
                current = current.get_left()
            else:
                current = current.get_right()
        return False
    
    def in_order(self, node):
        if node is not None:
            self.in_order(node.get_left())
            print(node.get_data())
            self.in_order(node.get_right())
        
```

### Q1 work — written (1d)


TreeNode is just a node as it is a child of the Node class. It has similar behaviour and characteristics, with an additional left side pointer. The extra pointer allows for the node to point to 2 different paths, allowing one to traverse the graph following a logic akin to binary search. This allows for significantly faster traversals at O(log n) just like binary search, compared to O(n) of a linear linked list.

### Q2 work — HashTable + driver


```python
class HashTable:
    #all member IDs are integers
    def __init__(self, N):
        self.__n = N
        self.__arr = []
        for i in range(N):
            self.__arr.append(LinkedList())
        
    def hash_function(self, key):
        return key % self.__n
    
    def insert(self, key):
        self.__arr[self.hash_function(key)].insert_front(key)
        
    def search(self, key):
        return self.__arr[self.hash_function(key)].search(key)
    
    def delete(self, key):
        self.__arr[self.hash_function(key)].delete(key)
        
    def display(self):
        for i in self.__arr:
            print(f"linked list object at {i} {i.display()}")
```

```python
testis = HashTable(7)
for i in [50, 20, 70, 10, 40, 57, 15]:
    testis.insert(i)
testis.display()
testis.search(40)
testis.search(99)
testis.delete(20)
print("\n")
testis.display()
```

### Q2 work — written (2c)


average search is O(1) as hash function is O(1), and on average cases in scenarios where the hash function hashes the key uniquely, the first index of the bucket is the key and the key is immediately found. However, if every single key is hashed to the same bucket, everything is chained into the singular linked list and average search is now O(n) which is search time in a linked list due to linear traversal.

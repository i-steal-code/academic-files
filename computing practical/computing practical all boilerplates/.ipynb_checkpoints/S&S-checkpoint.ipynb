linear search: probe everything in order along the line, 1 by 1 (hence linear)

time complexity: for as long as you sweep the list. best case is O(1), found on first term. worse case is O(n), found on last term

space complexity: O(1)

```python
def LinearSearch(Arr, FindValue):
    for i in range(len(Arr)):
        if Arr[i] == FindValue:
            return i  # return index of first match
    return -1  # not found
```

binary search: only works on sorted lists, but is way faster than linear search. takes findvalue, compares it to the middle and then moves to look at either left or right side of the list through a comparison to the middle. this halves the searchin scope for every pass, until it reaches the found target. 

time complexity: O(log n)

space complexity: O(1) for iterative, O(log n) for recursive

```python
def BinarySearch(Arr, FindValue, Low, High):
    middle = (Low + High)//2
    if Low > High:
        return -1
    elif Arr[middle] == FindValue:
        return middle
    elif FindValue < Arr[middle]:
        return BinarySearch(Arr,FindValue,Low,middle-1)
    else:
        return BinarySearch(Arr,FindValue,middle,High+1)
```

hash table search: create a hash table and use hash function to do an index lookup in the hash table.

time complexity: O(1), which is as fast as the hash function computes

space complexity: O(1) as no extra variables required

(not a very exam sanctioned version, for more comprehensive hash table code look at the ADT notebook)


```python
#hashing with separate chaining (LLM written, in early 2025)
BUCKET_SIZE = 7
class Hash(object):
    def __init__(self, bucket):
        # Number of buckets
        self.__bucket = bucket
        # Hash table of size bucket
        self.__table = [[] for _ in range(bucket)]

    # hash function to map values to key
    def hashFunction(self, key):
        return (key % self.__bucket)

    def insertItem(self, key):
        # get the hash index of key
        index = self.hashFunction(key)
        self.__table[index].append(key)

    def deleteItem(self, key):
        # get the hash index of key
        index = self.hashFunction(key)

        # Check the key in the hash table
        if key not in self.__table[index]:
            return

        # delete the key from hash table
        self.__table[index].remove(key)

    # function to display hash table
    def displayHash(self):
        for i in range(self.__bucket):
            print("[%d]" % i, end='')
            for x in self.__table[i]:
                print(" --> %d" % x, end='')
            print()


# Drive Program
if __name__ == "__main__":
    # array that contains keys to be mapped
    reset_test_list()

    # Create a empty hash table of BUCKET_SIZE
    h = Hash(BUCKET_SIZE)

    # insert the keys into the hash table
    for x in test_list:
        h.insertItem(x)

    # delete 12 from the hash table
    h.deleteItem(x)
    # Display the hash table
    h.displayHash()
```

insertion sort: front of list is sorted portion and back is unsorted. take objects in the list and compare to the sorted portion, looking for an insertion point where the left neighbour is smaller and right neighbour is larger. this sweeps the list to insert an object for every object until the size of the sorted portion is the size of the list.


time complexity: O(n^2) 

space complexity: in-place sorting O(1) 

```python
def InsertionSort(Arr):
    for i in range(1, len(Arr)):
        current = Arr[i]  # object taken from unsorted portion
        j = i - 1
        # shift larger sorted objects right until insertion spot found
        while j >= 0 and Arr[j] > current:
            Arr[j + 1] = Arr[j]
            j -= 1
        Arr[j + 1] = current  # insert into sorted portion
    return Arr
```

bubble sort: on every pass, look at current object and next object. swap to get them in the correct order, and move on to next index. keep passing until the final pass returns 0 swaps, which is when you know the whole list is sorted.

time complexity: O(n^2)

space complexity: in-place O(1)

```python
def BubbleSort(Arr):
    n = len(Arr)
    swapped = True
    while swapped:
        swapped = False
        for i in range(n - 1):
            if Arr[i] > Arr[i + 1]:
                # swap to put current pair in correct order
                temp = Arr[i]
                Arr[i] = Arr[i + 1]
                Arr[i + 1] = temp
                swapped = True
        # after a no-swap pass, list is sorted
    return Arr
```

quicksort: a divide and conquer algorithm. recursively partition parts of the list to conquer each part of the list bit by bit. 

partition: take pivotvalue out, and move the leftmark index until its larger than the pivot. then move rightmark index until it is smaller than pivot. you swap the objects at leftmark and rightmark, and then continue partitioning on the array encompassed by the current left and right marks. this shrinks the partitioning array until you end up partitioning nothing (bounds cross) and we conclude this recurse of partitions by acknowledging that all terms on the right of the rightmark are larger than the pivot and everything else on the left is smaller. 

conquer: recursively quicksort the left sub-array, which partitions and partitions and gives you a group of objects larger than the previous pivot in an ascending order. because of this, pivots are already in their final sorted positions after the partition. once a sub-array becomes 1 object, the 1 object is the pivot for that partition recursion and hence is sorted. after all left sub-arrays have been quicksorted, we move on to quicksort the right side using the same logic. 

quicksort is recursive, with the algorithm sorting to the left then to the right to sweep through everything. 

time complexity: O(n log n) but O(n^2) as a worse case

space complexity: in-place data is O(1) extra array space, but recursion depth is O(log n) average and O(n) worst

```python
def Partition(Arr, First, Last):
  pivotValue = Arr[First]
  leftmark = First + 1 #index of second left object
  rightmark = Last #index of right-most object
  done = False

  while not done:
    while leftmark <= rightmark and Arr[leftmark] <= pivotValue:
      #increment leftmark index until leftmark array value >= first term
      leftmark += 1
    while Arr[rightmark] >= pivotValue and rightmark >= leftmark:
      #decrease rightmark index until rightmark array value <= first term
      rightmark -= 1
    #the loops above will stop if rightmark approaches leftmark

    if rightmark < leftmark:
      #end the loop when rightmark index < leftmark index
      done = True
    else:
      #swap leftmark and rightmark value in array
      temp = Arr[leftmark]
      Arr[leftmark] = Arr[rightmark]
      Arr[rightmark] = temp
  #swap first term and rightmark value
  temp = Arr[First]
  Arr[First] = Arr[rightmark]
  Arr[rightmark] = temp
  return rightmark #output rightmark index

def Quicksort(Array, First, Last):
  if First < Last:
    SplitPoint = Partition(Array, First, Last)
    #recursively quicksort left half
    Quicksort(Array, First, SplitPoint - 1)
    #recursively quicksort right half
    Quicksort(Array, SplitPoint + 1, Last)
  return Array #return once fully sorted where index term >= last index
```

mergesort: divide and conquer. (split into binary tree, sort with merging)

divide:
takes list and splits in half until you get single object arrays (which are considered sorted). from there, you take your sorted arrays and are ready to conquer

conquer:
merge the 2 selected arrays together by comparing their 0th index. smaller number gets popped and appended into the merged array until both lists are empty. tada! you have an merged (and sorted) array.

since mergesort is recursive (usually), the merging of the binary tree is done as you exit the recursion loops and you end up with 1 big merged and sorted array. 

time complexity: O(n log n) 

space complexity: not in-place so O(n)

```python
def merge(array, low, mid, high):
  left = array[low : mid + 1]
  right = array[mid + 1 : high + 1]
  i = 0
  j = 0
  k = low
  while i < len(left) and j < len(right):
    if left[i] <= right[j]:
      array[k] = left[i]
      i += 1
    else:
      array[k] = right[j]
      j += 1
    k += 1
  while i < len(left):
    array[k] = left[i]
    i += 1
    k += 1
  while j < len(right):
    array[k] = right[j]
    j += 1
    k += 1
  return array

def mergeSort(array, low, high):
  if low < high:
    mid = (low + high) // 2
    mergeSort(array, low, mid)
    mergeSort(array, mid + 1, high)
    merge(array, low, mid, high)
  return array
```

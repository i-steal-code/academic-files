## 1. Relational theory

### Vocabulary
| Term | Meaning |
|------|---------|
| Table / relation | 2D set of records with named fields |
| Record / tuple | one row |
| Field / attribute | one column |
| Primary key (PK) | uniquely identifies a record; not null |
| Foreign key (FK) | field that references another table's PK |
| Composite key | PK made of **two or more** fields together |
| Secondary key | not unique, used for searching (e.g. class, name) |

### Normalisation (exam recall)
- **1NF:** atomic values, no repeating groups
- **2NF:** 1NF + no partial dependence on part of a composite PK
- **3NF:** 2NF + no transitive dependence (non-key field depends only on the PK)

JOINs put split-3NF tables back together for a query. They do **not** “revert the database to 1NF”; the stored tables stay normalised.

### SQL vs NoSQL (one-liners)
- SQL: fixed schema, strong consistency, JOINs, good for related tables
- NoSQL (document): flexible schema, nested arrays/objects, scales out; “joins” are often embedding or a second query

## 2. SQLite boilerplates

Exam loop: **connect → DROP/CREATE → INSERT from file → SELECT/UPDATE → commit → close**.

Use **`?` placeholders** for values. `INSERT OR IGNORE` skips PK clashes. `fetchone()` / `fetchall()` pull results.

**PK trap:** if a student can be late on many dates, `Late.stu_id` cannot be the only PK. Use composite `(date, stu_id)` (or a surrogate `late_id`).


````python
import sqlite3, csv

conn = sqlite3.connect("college.db")
conn.row_factory = sqlite3.Row
cur = conn.cursor()


cur.execute("DROP TABLE IF EXISTS Late")
cur.execute("DROP TABLE IF EXISTS Student")

cur.execute("""
CREATE TABLE IF NOT EXISTS Student(
    stu_id INTEGER PRIMARY KEY,
    name TEXT,
    civics_class TEXT,
    handphone INTEGER
)
""")
cur.execute("""
CREATE TABLE IF NOT EXISTS Late(
    date TEXT,
    stu_id INTEGER,
    reason TEXT,
    PRIMARY KEY (date, stu_id),
    FOREIGN KEY (stu_id) REFERENCES Student(stu_id)
)
""")

with open("STUDENT.csv") as file:
    reader = csv.reader(file)
    next(reader)  # skip header
    for row in reader:
        cur.execute("""
            INSERT OR IGNORE INTO Student(stu_id, name, civics_class, handphone)
            VALUES (?,?,?,?)
        """, (row[0], row[1], row[2], row[3]))

with open("LATE.csv") as file:
    reader = csv.reader(file)
    next(reader)
    for row in reader:
        cur.execute("""
            INSERT OR IGNORE INTO Late(date, stu_id, reason)
            VALUES (?,?,?)
        """, (row[0], row[1], row[2]))

conn.commit()
print("tables created and seeded")
````

### CRUD template


````python
# CRUD pattern (parameterised — never splice user input into SQL)
# CREATE
cur.execute("INSERT INTO Student(stu_id, name, civics_class, handphone) VALUES (?,?,?,?)",
            (25999, "Test Student", "25S02X", 81234567))

# READ
cur.execute("SELECT name, civics_class FROM Student WHERE stu_id = ?", (25999,))
for i in cur.fetchone():
    print(i)

# UPDATE
cur.execute("UPDATE Student SET name = ? WHERE stu_id = ?", ("Renamed Student", 25999))

# DELETE
cur.execute("DELETE FROM Student WHERE stu_id = ?", (25999,))
conn.commit()
````

## 3. SQL query lookup

Pattern:

```sql
SELECT fields / aggregates
FROM table
     [INNER | LEFT OUTER] JOIN other ON pk = fk
WHERE condition          -- filter rows before grouping
GROUP BY field           -- one summary row per group
HAVING aggregate_cond    -- filter groups
ORDER BY field [DESC]
```

`LIKE`: `name LIKE ?` with `"%Ali%"` (wildcards around the search term).

INNER JOIN drops unmatched rows. LEFT OUTER JOIN keeps every left-table row; missing right side comes back as `NULL`.


````python
# INNER JOIN: only students who have at least one late record
cur.execute("""
SELECT Student.name, Late.date, Late.reason
FROM Student INNER JOIN Late ON Student.stu_id = Late.stu_id
WHERE Student.civics_class = ?
ORDER BY Late.date
""", ("25S02X",))
#fetch all records in dictionary form from selection filter
for row in cur.fetchall():
    print(row['name'], row['date'], row['reason'])

# LEFT OUTER JOIN: keep students even if they have zero late records
cur.execute("""
SELECT Student.name, Late.date
FROM Student LEFT OUTER JOIN Late ON Student.stu_id = Late.stu_id
WHERE Student.civics_class = ?
""", ("25S02X",))
# names with date = None were never late
````

### Exam task B — GROUP BY / HAVING / LIKE

**(a)** Write SQL to list each `civics_class` and **how many late records** that class has.  
Hint: JOIN, then `GROUP BY` class, `COUNT(*)`.

**(b)** Same as (a), but only classes with **at least 3** late records (`HAVING`).

**(c)** Names of students whose name contains `"Li"` (`LIKE`).

**(d)** In 2–3 sentences: what does `GROUP BY` do to rows, and why can you not `SELECT Late.reason` in the same query without aggregating it?

Write SQL in the cell (run against `college.db` after section 2).


````python
cur.execute("""
SELECT student.civics_class, COUNT(*) AS total 
FROM Student INNER JOIN Late ON Student.stu_id = Late.stu_id
GROUP BY Student.civics_class;
""")
for row in cur.fetchall():
    print(row['civics_class'], row['total'])
````

````python
cur.execute("""
SELECT student.civics_class, COUNT(*) AS total 
FROM Student INNER JOIN Late ON Student.stu_id = Late.stu_id
GROUP BY Student.civics_class 
HAVING total >= 3;
""")
for row in cur.fetchall():
    print(row['civics_class'], row['total'])
````

````python
cur.execute("""
SELECT name 
FROM Student
WHERE name LIKE "%Li%"
""")
for row in cur.fetchall():
    print(row['name'])
````

GROUP BY groups records by a certain field, giving you a list of records that are groups of records with the same value in a particular field. this allows you to apply a filter on aggregate functions for a specific value. you can no longer SELECT Late.reason as the new records are groups that have different "reason" fields.

## 5. NoSQL + pymongo

Document model: **database → collection → document (dict)**. Fields can be nested / arrays — that is why embedding can replace a SQL JOIN.

Exam-typical pymongo:
```text
client = MongoClient("localhost", 27017)   # or mongodb://localhost:27017/
db = client["name"]
coll = db["name"]
coll.delete_many({})
coll.insert_one(doc) / insert_many(list)
coll.find(filter, projection)
coll.find_one(filter)
coll.update_one(filter, {"$set": {...}, "$inc": {...}})
coll.delete_one(filter)
coll.drop()   # or client.drop_database("name")
```

**Join in exams:** usually a second `find` / `$in` list of IDs, **or** embed related data inside one document. Aggregation `$lookup` is extra, not required for 9569 practicals.


````python
import pymongo, csv, json

client = pymongo.MongoClient("mongodb://localhost:27017/")
db = client["college_db"]
students = db["student"]
lates = db["late"]
students.delete_many({})
lates.delete_many({})

with open("STUDENT.csv") as file:
    reader = csv.reader(file)
    headers = next(reader)
    for row in reader:
        students.insert_one(dict(zip(headers, row)))

with open("LATE.csv") as file:
    reader = csv.reader(file)
    headers = next(reader)
    for row in reader:
        lates.insert_one(dict(zip(headers, row)))

# find / projection / sort
for doc in students.find({"Class_Group": "25S02X"}, {"_id": 0, "Student_Name": 1, "Student_ID": 1}):
    print(doc)

# two-query "join" (exam-typical; no aggregation pipeline needed)
late_ids = lates.distinct("Student_ID")
for doc in students.find({"Student_ID": {"$in": late_ids}}, {"_id": 0, "Student_Name": 1}):
    print(doc["Student_Name"])

client.close()
````

### Exam task D — find, update, embed

**(a)** `find` all late documents with `Date` greater than `"2024-07-02"`. Project away `_id`. Print `Student_ID` and `Reason`.

**(b)** `update_one`: set `Reason` to `"Updated reason"` for student `"25065"` on `"2024-07-03"`.

**(c)** Design choice: store late records as an **array embedded** in each student document vs two collections. When is embedding simpler, and when does it become a bad idea? (2–4 sentences)

**(d)** Write Python to rebuild **embedded** student documents: each student dict also has `"lates": [ {date, reason}, ... ]`. Insert into a new collection `student_embed`. Then query students who have **at least one** late record without using `$lookup`.


````python
# Task D — needs MongoDB running
lates.find({"Date": {"$gt": "2024-07-02"}}, {"_id": 0, "Student_ID": 1, "Reason": 1})
    
lates.update_one({"Reason": "Updated reason"}, {"$set": {"Student_ID": "25065", "Date": "2024-07-03"}})
````

embedding is simpler when the numbers of layers of embedding is small, allowing for easier fetching of data without having to join 2 documents together. embedding becomes unfeasible when the number of separate documents is large and multiple layers of embedding is needed, complicating the database more than if all the documents were separated.

**(b)** Map these SQL ideas to Mongo: table, record, field, primary key, JOIN.

**(c)** Why is `row_factory = sqlite3.Row` optional for exams? When would it help?


a relational database requires a fixed schema, and in store addresses, records can take on many different formats with different fields. relational database will not be able to accomodate for the flexible nature of store addresses. SQL is a better fit if complex and varied queries on a multi-table relational database is required, such as when another table containing store tenants is added. 

SQL idea -> mongodb equivalent
table -> dictionary in database
record -> dictionary object in a 'table'
field -> value in the key-value pairs of each dictionary object ('records')
primary key -> id
JOIN -> find record where a particular field's value is "$in" another 'table' 

row_factory = sqlite3.row is optional for exams because it is not needed to complete most questions. it would be helpful when records fetched from the database need to be in dictionary form for indexing data using keys instead of numerical indexes that may be confusing. ## 7. Exam cheat-sheet

### sqlite3
```text
conn = sqlite3.connect("x.db"); cur = conn.cursor()
CREATE TABLE (... PK, FK REFERENCES other(pk))
INSERT OR IGNORE INTO t(...) VALUES (?,?,?)
SELECT ... FROM a INNER JOIN b ON a.pk = b.fk WHERE ... ORDER BY ...
SELECT f, COUNT(*) FROM ... GROUP BY f HAVING COUNT(*) >= ?
UPDATE t SET field=? WHERE pk=?
DELETE FROM t WHERE pk=?
conn.commit(); conn.close()
csv: next(reader) then loop rows
```

### pymongo
```text
coll.insert_one(dict) / insert_many(list)
coll.find({field: value}, {"_id":0, field:1})
coll.find({"n": {"$gte": 10}})
coll.update_one(filter, {"$set": {field: val}})
coll.delete_many({})
join: ids = coll.distinct("fk"); other.find({"pk": {"$in": ids}})
embed: parent["children"] = [ {...}, ... ]
```

### Theory
```text
PK unique+not null; FK references PK; composite = two+ fields
1NF atomic → 2NF no partial dep → 3NF no transitive dep
INNER = matches only; LEFT = keep all left rows
GROUP BY compresses groups; HAVING filters those groups
```

### Drill 1 — Binary search (15 min)
Write `BinarySearch(Arr, FindValue, Low, High)` recursive, sorted list only.

Test: `Arr = [3, 7, 12, 18, 22]` → search `12` returns index `2`, search `99` returns `-1`.

**Must have:** `Low > High` base case; recurse `middle - 1` / `middle + 1`.


````python
# Drill 1
def BinarySearch(Arr, FindValue, Low, High):
    middle = (High + Low) // 2
    if Low > High: #bounds have crossed; value not found in list
        return -1
    elif FindValue == Arr[middle]: #target found
        return middle
    elif FindValue > Arr[middle]: #move right
        BinarySearch(Arr, FindValue, middle, High)
    elif FindValue <= Arr[middle]: #move left
        BinarySearch(Arr, Findvalue, Low, middle)
    
nig = [3, 7, 12, 18, 22]
BinarySearch(nig, 12, 0, len(nig)-1)
# BinarySearch(nig, 99, 0, len(nig)-1)
````

### Drill 2 — Insertion sort one pass (10 min)
Given **already sorted prefix** `[2, 5, 9]` and unsorted tail starting with `4`, show the list **after one outer-loop iteration** of insertion sort (ascending). Then write full `InsertionSort(Arr)` and sort `[9, 4, 2, 7]`.

Expected final: `[2, 4, 7, 9]`.


````python
# Drill 2
#after 1 sorting loop iteration, it will be [4,2,5,9]

def InsertionSort(Arr):
    swap = True
    while swap:
        n = 0
        last = Arr.pop()
        try:
            while Arr[n] < last:
                n += 1
        except:
            #breaks out of the loop when index of item larger than last goes off the list; last is already largest in the list
            #this is now a sorted case
            swap = False
        #n here is the index of the item in the array that is just larger than last
        Arr.insert(n, last)
    return Arr
        
test_list = [9,4,2,7]
InsertionSort(test_list)
````

### Drill 5 — Linked list delete (15 min)
From memory: `Node` + `LinkedList` with `insert_end`, `search`, `delete`.

Driver: insert `50, 20, 70, 10, 40` → delete `20` → delete `99` (no crash) → delete head `50`.

**Trap to pass:** `delete` when not found must return without error.


````python
# Drill 5

class Node:
    def __init__(self, data = None):
        self.__data = data
        self.__pointer = None
        
    def get_data(self):
        return self.__data
    def set_data(self, data):
        self.__data = data
        
    def get_pointer(self):
        return self.__pointer
    def set_pointer(self, pointer):
        self.__pointer = pointer
        
class LinkedList:
    def __init__(self, start = Node()): #start pointer is a start node
        self.__start = start
        
    def insert_end(self, data):
        #insert into start node if it is empty
        if self.__start.get_data() is None:
            self.__start.set_data(data)
            return 
        #traverse until current node is null
        current = self.__start
        while current: 
            previous = current
            current = current.get_pointer()
        #append a node onto the previous node pointer
        previous.set_pointer(Node(data))
        return 
    
    def search(self, data):
        #this function becomes traverse in the event that a question asks for a search function that explicitly only outputs true/false or current node only
        previous = None
        current = self.__start
        while current:
            if current.get_data() == data:
                return previous, current
            else:
                previous = current
                current = current.get_pointer()
        return None
                
    def delete(self, data):
        find = self.search(data)
        #for self reference: previous = find[0], current = find[1]
        try:
            find[0].set_pointer(find[1].get_pointer())
        except:
            return None
        
    def display(self):
        previous = None
        display = ''
        current = self.__start
        while current:
            display += f"-> {current.get_data()}"
            current = current.get_pointer()
        return display
````

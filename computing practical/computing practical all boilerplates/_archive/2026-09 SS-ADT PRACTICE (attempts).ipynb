### Drill 2 — Insertion sort one pass (10 min)
Given **already sorted prefix** `[2, 5, 9]` and unsorted tail starting with `4`, show the list **after one outer-loop iteration** of insertion sort (ascending). Then write full `InsertionSort(Arr)` and sort `[9, 4, 2, 7]`.

Expected final: `[2, 4, 7, 9]`.

````python
# Drill 2  — YOUR ATTEMPT (preserved)
#after 1 sorting loop iteration, it will be [4,2,5,9]

def InsertionSort(Arr):
    swap = True
    while swap:
        n = 0
        last = Arr.pop()
        try:
            while Arr[n] < last:
                n += 1
        except:
            #breaks out of the loop when index of item larger than last goes off the list; last is already largest in the list
            #this is now a sorted case
            swap = False
        #n here is the index of the item in the array that is just larger than last
        Arr.insert(n, last)
    return Arr
        
test_list = [9,4,2,7]
InsertionSort(test_list)
````

#### Review — Drill 2

**Your trace answer `[4, 2, 5, 9]` is wrong.** With sorted prefix `[2, 5, 9]` and next item `4`:
`9 > 4` shift → `5 > 4` shift → `2 ≤ 4` stop → insert at index 1 → **`[2, 4, 5, 9]`**.

**Your algorithm is genuinely broken — this is measured, not a style opinion.**
Ran it against every permutation of length 2–6 plus 3000 random lists with duplicates:

> **2067 failures out of 3872 lists (53%).**
> Smallest counterexample: `[2, 1, 3]` → returns **`[2, 1, 3]`**, unsorted.

**Why it fails:** you `pop()` the *last* element and insert it into the prefix as if the prefix were already sorted — it is not. Termination hangs on an `IndexError` firing when the popped item happens to be the largest, which can occur while the rest is still out of order. `[9, 4, 2, 7]` passing was luck.

**The real shape:** walk **forward** `i = 1 … n-1`, hold `Arr[i]`, shift larger items right with a backwards `j`.

````python
# Drill 2 — corrected reference
def InsertionSort(Arr):
    for i in range(1, len(Arr)):
        current = Arr[i]              # next unsorted item
        j = i - 1
        while j >= 0 and Arr[j] > current:
            Arr[j + 1] = Arr[j]       # shift larger item right
            j -= 1
        Arr[j + 1] = current          # drop into its slot
    return Arr

print(InsertionSort([9, 4, 2, 7]))    # [2, 4, 7, 9]
print(InsertionSort([2, 1, 3]))       # [1, 2, 3]  <- your version returned [2, 1, 3]
````

**Bug 1 — head delete silently does nothing (this is the drill's trap).**
`search` returns `(previous, current)` with `previous = None` for a head match. `find[0].set_pointer(...)` then raises `AttributeError`, your bare `except` swallows it, and the head survives. A bare `except` that hides a real failure is the dangerous part — the same `except` is also what makes the not-found case "work".

**Bug 2 — mutable default argument.** `def __init__(self, start = Node())` evaluates `Node()` **once**, at definition time, so every `LinkedList` shares one node. Verified: built list `x`, then a fresh list `y`, and `y` already contained `x`'s data.

`insert_end`, `search` and `display` are otherwise sound.

````python
# Drill 5 — corrected reference
class Node:
    def __init__(self, data=None):
        self.__data = data
        self.__right = None
    def get_data(self):      return self.__data
    def set_data(self, d):   self.__data = d
    def get_right(self):     return self.__right
    def set_right(self, r):  self.__right = r


class LinkedList:
    def __init__(self):
        self.__start = None                      # None = empty, no sentinel node

    def insert_end(self, data):
        new = Node(data)
        if self.__start is None:
            self.__start = new
            return
        current = self.__start
        while current.get_right() is not None:
            current = current.get_right()
        current.set_right(new)

    def search(self, data):
        current = self.__start
        while current is not None and current.get_data() != data:
            current = current.get_right()
        return current                            # Node or None

    def delete(self, data):
        if self.__start is None:
            return
        if self.__start.get_data() == data:       # HEAD CASE - the one you missed
            self.__start = self.__start.get_right()
            return
        previous, current = None, self.__start
        while current is not None and current.get_data() != data:
            previous = current
            current = current.get_right()
        if current is None:
            return                                # not found, no error
        previous.set_right(current.get_right())

    def display(self):
        out, current = "", self.__start
        while current:
            out += f" -> {current.get_data()}"
            current = current.get_right()
        return out


ll = LinkedList()
for v in [50, 20, 70, 10, 40]:
    ll.insert_end(v)
print("initial:    ", ll.display())
ll.delete(20);  print("delete 20:  ", ll.display())
ll.delete(99);  print("delete 99:  ", ll.display())
ll.delete(50);  print("delete head:", ll.display())
````

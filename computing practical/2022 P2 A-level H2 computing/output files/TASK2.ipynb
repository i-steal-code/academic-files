```python
#Task 2.1

import random

def task2_1():
    output = []
    for i in range(100):
        output.append(int(random.randint(1,100)))
    return output

print(task2_1())
```

```python
#Task 2.2

def bubblesort(arr):
    length = len(arr)
    swapped = True
    while swapped:
        swapped = False
        for n in range(length-1):
            if arr[n] > arr[n+1]:
                temp = arr[n]
                arr[n] = arr[n+1]
                arr[n+1] = temp
                swapped = True
    return arr

print(bubblesort(task2_1()))
```

```python
#Task 2.3

def merge(arr, low, high):
    mid = (low + high) // 2
    left = arr[low : mid + 1]
    right = arr[mid + 1 : high + 1]
    l = 0
    r = 0
    a = low
    #write onto arr whichever 
    while l < len(left) and r < len(right):
        if left[l] <= right[r]:
            arr[a] = left[l]
            l += 1
            a += 1
        else:
            arr[a] = right[r]
            r += 1
            a += 1
    #leftovers
    while l < len(left):
        arr[a] = left[l]
        l += 1
        a += 1
    while r < len(right):
        arr[a] = right[r]
        r += 1
        a += 1
    return arr
    
    
def mergesort(arr, low, high):
    if low < high:
        mid = (low + high) // 2
        mergesort(arr, low, mid)
        mergesort(arr, mid + 1, high)
        merge(arr, low, high)
    return arr

#menu
valid = False
while not valid:
    what = input("which sorting algorithm do you want to use? algorithm: ").lower()
    if what in ['bubblesort', 'mergesort','bubble','merge']:
        valid = True
    else:
        print('invalid input, please try again')
test = task2_1()
if what in ['bubblesort','bubble']:
    print(bubblesort(test))
else:
    print(mergesort(test,0,99))
```

```python
#Task 2.4

def binarysearch(arr, low, high, value):
    mid = (low + high) // 2
    if arr[mid] == value:
        return mid
    elif low >= high:
        return -1
    elif value < int(arr[mid]):
        return binarysearch(arr, low, mid-1, value)
    elif value > int(arr[mid]):
        return binarysearch(arr, mid+1, high, value)

test = bubblesort(task2_1())
print(f'the test list we will be using is {test}')
for i in range(2):
    find = int(input('key in a value to search for: '))
    index = binarysearch(test, 0, 99, find)
    if index != -1:
        print(f'result: index {index}')
    else:
        print('not found')
```

```python
#Task 1.1

def encrypt(char):
    #validation checks
    if len(char) != 1 or not char.isascii():
        return -1
    #conversion to ascii code
    value = ord(char)
    #exclude non-letter or space characters
    if (value >= 65 and value <= 90) or (value >= 97 and value <= 122) or value == 32:
        pass
    else:
        return -1
    #convert spaces to exclamation mark
    if value == 32:
        return '!'
    value += 10
    #tame overflows for upper and lower case letters
    if value > 90 and value <= 100: #if char is upper case letter overflowing
        overflow = value-90
        value = 65 + overflow - 1
    elif value > 122 and value <= 132: #if char is lower case letter overflowing
        overflow = value-122
        value = 97 + overflow - 1
    return chr(value).strip()
```

```python
#Task 1.2

for _ in range(4):
    read = input("please enter 1 character to encrypt: ")
    if len(read) != 1:
        print('invalid length')
    elif not read.isascii():
        print('non-ascii character input')
    print(encrypt(read))
```

```python
#Task 1.3

with open('DATATOENCRYPT.txt') as file:
    work = ''
    for row in file:
        work += row
    #encrypt every char
    work = [encrypt(_) for _ in work]
    #piece together everything
    collector = ''
    for i in work:
        if i != -1:
            collector += i

with open('ENCRYPTEDMESSAGE.txt', 'w') as file:
    #write onto output file
    file.write(collector)
```

```python
#Task 4.1

#init
import sqlite3
conn = sqlite3.connect('Task4_4/LIBRARY.db')
conn.row_factory = sqlite3.Row
cur = conn.cursor()

#reset db
cur.execute("""
DROP TABLE IF EXISTS Book
""")
cur.execute("""
DROP TABLE IF EXISTS Member
""")
cur.execute("""
DROP TABLE IF EXISTS Loan
""")

#schema
cur.execute("""
CREATE TABLE Book(
BookID INTEGER PRIMARY KEY,
Title TEXT,
Genre TEXT
)
""")
cur.execute("""
CREATE TABLE Member(
MemberNumber INTEGER PRIMARY KEY,
FamilyName TEXT,
GivenName TEXT
)
""")
cur.execute("""
CREATE TABLE Loan(
LoanID INTEGER PRIMARY KEY,
MemberNumber INTEGER,
BookID INTEGER,
DateLoaned TEXT,
Returned TEXT,
FOREIGN KEY(MemberNumber) REFERENCES Member,
FOREIGN KEY(BookID) REFERENCES Book
)
""")
```

```python
#Task 4.2

#seed
with open('BOOK.txt') as file:
    for row in file:
        row = row.strip().split(',')
        cur.execute("""
        INSERT OR IGNORE INTO Book (BookID, Title, Genre)
        VALUES (?,?,?)
        """, (row[0],row[1],row[2]))
with open('MEMBER.txt') as file:
    for row in file:
        row = row.strip().split(',')
        cur.execute("""
        INSERT OR IGNORE INTO Member (MemberNumber, FamilyName, GivenName)
        VALUES (?,?,?)
        """, (row[0],row[1],row[2]))
with open('LOAN.txt') as file:
    for row in file:
        row = row.strip().split(',')
        cur.execute("""
        INSERT OR IGNORE INTO Loan (LoanID,MemberNumber,BookID,Dateloaned,Returned)
        VALUES (?,?,?,?,?)
        """, (row[0],row[1],row[2],row[3],row[4]))
conn.commit()
```

```python
#Task 4.3

num = input('enter member number to search for: ')
cur.execute("""
SELECT Book.Title, Loan.Returned
FROM Book INNER JOIN Loan ON Book.BookID = Loan.BookID
WHERE Loan.MemberNumber = ?
""", (int(num.strip()),))
results = cur.fetchall()
if results:
    for row in results:
        if row['Returned'] == 'TRUE':
            returned = 'returned'
        else:
            returned = 'not returned'
        print(f"{row['Title']} | {returned}")
else:
    print('no matching member number found')
```

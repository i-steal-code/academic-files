```python
#Task 3.1

import random

class Tree:
    def __init__(self, data = None, left = None, right = None):
        self.__left_pointer = left
        self.__right_pointer = right
        self.__data = data
        
    def get_data(self):
        return self.__data
    
    def set_data(self, data):
        self.__data = data
        
    def get_left(self):
        return self.__left_pointer
    
    def set_left(self, left):
        self.__left_pointer = left
        
    def get_right(self):
        return self.__right_pointer
    
    def set_right(self, right):
        self.__right_pointer = right
        
    def insert(self, current, newdata):
        #empty node
        if current is None: 
            return None
        #empty head node 
        elif current.get_data() is None:
            current.set_data(newdata)
            return True
        #regular cases
        elif newdata <= current.get_data(): #smaller than or equal current - go left (data seniority)
            #insert into left if there is nothing left
            if current.get_left() is None:
                current.set_left(Tree(newdata))
                return True
            #if not, traverse to left node as per normal and recurse
            return self.insert(current.get_left(), newdata)
        else: #larger than current - go right
            #insert into right if there is nothing right
            if current.get_right() is None:
                current.set_right(Tree(newdata))
                return True
            #else traverse right
            return self.insert(current.get_right(), newdata)
            
        
    def in_order(self, current):
        if current:
            self.in_order(current.get_left())
            print(current.get_data())
            self.in_order(current.get_right())
        else:
            return None
        
    def post_order(self, current):
        if current:
            self.post_order(current.get_left())
            self.post_order(current.get_right())
            print(current.get_data())
        else:
            return None
        
#main
if __name__ == "__main__":
    test = Tree()
    values = [random.randint(0,999) for _ in range(10)]
    print(values)
    for item in values:
        test.insert(test,item)
    print('inorder traversal')
    test.in_order(test)
    print('postorder traversal')
    test.post_order(test)
```

```python
#Task 2.1

class Person:
    def __init__(self,name,age):
        self.__name = name
        self.__age = int(age)
        
    def get_data(self, typ = None):
        if typ == 'name':
            return self.__name
        elif typ == 'age':
            return self.__age
        else:
            return self.__name, self.__age
    
    def print(self):
        print(self.__name, self.__age)
```

```python
#Task 2.2

def task2_2(filename):
    with open(filename) as file:
        output = []
        for row in file:
            row = row.split(',')
            output.append(Person(row[0],row[1][:-1]))
    return output
            
list_of_person = task2_2('PERSON.txt')
for person in list_of_person:
    person.print()
```

```python
#Task 2.3

def task2_3(list_of_person,key,order):
    for n in range(1, len(list_of_person)):
        current = list_of_person[n] 
        i = n - 1
        while i >= 0 and list_of_person[i].get_data(key) > current.get_data(key):
            list_of_person[i+1] = list_of_person[i]
            i -= 1
        list_of_person[i+1] = current
    if order == 'desc':
        list_of_person.reverse()
    return list_of_person

list_of_person = task2_2('PERSON.txt')
task2_3(list_of_person, 'name', 'asc')
for person in list_of_person:
    person.print()
```

```python
#Task 2.4

def partition(arr, first, last, key):
    pivot = arr[first]
    leftmark = first + 1
    rightmark = last
    done = False
    while not done:
        while leftmark <= rightmark and arr[leftmark].get_data(key) <= pivot.get_data(key):
            leftmark += 1
        while rightmark >= leftmark and arr[rightmark].get_data(key) >= pivot.get_data(key):
            rightmark -= 1
        if leftmark > rightmark:
            done = True
        else: 
            temp = arr[leftmark]
            arr[leftmark] = arr[rightmark]
            arr[rightmark] = temp
    arr[first] = arr[rightmark]
    arr[rightmark] = pivot
    return rightmark

def quicksort(arr, first, last, key):
    if first < last:
        split = partition(arr, first, last, key)
        quicksort(arr, first, split - 1, key)
        quicksort(arr, split + 1, last, key)
    return arr

def task2_4(list_of_person,key,order):
    output = quicksort(list_of_person, 0, len(list_of_person)-1, key)
    if order == 'desc':
        output.reverse()
    return output

list_of_person = task2_2('PERSON.txt')
task2_4(list_of_person, 'age', 'desc')
for person in list_of_person:
    person.print() 
```

```python
#Task 2.5

def task2_5(list_of_person, method, key, order):
    if method == 'quick sort':
        return task2_4(list_of_person, key, order)
    elif method == 'insertion sort':
        return task2_3(list_of_person, key, order)
    
list_of_person = task2_2('PERSON.txt')
task2_5(list_of_person, 'quick sort', 'name', 'desc')
for person in list_of_person:
    person.print() 
```

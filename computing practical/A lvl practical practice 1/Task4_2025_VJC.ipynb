```python
#Task 4.1

import sqlite3

conn = sqlite3.connect('Task_4_4/BUSROUTES.db')
conn.row_factory = sqlite3.Row
cur = conn.cursor()

#schema
cur.execute("""
DROP TABLE IF EXISTS Service
""")
cur.execute("""
DROP TABLE IF EXISTS Stop
""")
cur.execute("""
DROP TABLE IF EXISTS Route
""")
cur.execute("""
CREATE TABLE Service(
ServiceNo TEXT PRIMARY KEY,
Operator TEXT,
Category TEXT
)
""")
cur.execute("""
CREATE TABLE Stop(
RoadName TEXT,
Description TEXT,
BusStopCode INTEGER PRIMARY KEY
)
""")
cur.execute("""
CREATE TABLE Route(
ServiceNo TEXT,
Operator TEXT,
Direction INTEGER,
StopSequence INTEGER,
BusStopCode INTEGER,
PRIMARY KEY (ServiceNo, BusStopCode),
FOREIGN KEY (ServiceNo) REFERENCES Service,
FOREIGN KEY (BusStopCode) REFERENCES Stop
)
""")
```

```python
#Task 4.2

#seed

import csv

with open('Service.csv') as f:
    file = csv.reader(f)
    next(file)
    for row in file:
        cur.execute("""
        INSERT OR IGNORE INTO Service (ServiceNo, Operator, Category)
        VALUES (?,?,?)
        """, (row[0], row[1], row[2]))
        
with open('Stop.csv') as f:
    file = csv.reader(f)
    next(file)
    for row in file:
        cur.execute("""
        INSERT OR IGNORE INTO Stop (BusStopCode, RoadName, Description)
        VALUES (?,?,?)
        """, (row[0], row[1], row[2]))
        
with open('Route.csv') as f:
    file = csv.reader(f)
    next(file)
    for row in file:
        cur.execute("""
        INSERT OR IGNORE INTO Route (ServiceNo,Operator,Direction,StopSequence,BusStopCode)
        VALUES (?,?,?,?,?)
        """, (row[0], row[1], row[2], row[3], row[4]))
        
conn.commit()
```

```python
#Task 4.3

def get_routes(ServiceNo):
    cur.execute("""
    SELECT Route.Direction, Route.BusStopCode, Stop.RoadName, Stop.Description
    FROM Route INNER JOIN Stop ON Route.BusStopCode = Stop.BusStopCode
    WHERE Route.ServiceNo = ?
    ORDER BY Route.Direction, Route.StopSequence ASC
    """, (ServiceNo.lower(),))
    for row in cur.fetchall():
        print(
            row['Direction'],
            row['BusStopCode'],
            row['RoadName'],
            row['Description']
        )
    return 

get_routes('10')
```

```python
#Task 3.1

class Person:
    def __init__(self, name, birthdate):
        self.name = name
        self.birthdate = birthdate
        
    def __repr__(self):
        return f"<{self.name}:{self.birthdate}>"
        
    def get_birthdate(self):
        return self.birthdate
    
    def get_name(self):
        return self.name
        
with open('birthdays.txt') as file:
    birthdays = []
    for row in file:
        row = row.split(',')
        birthdays.append(Person(row[0], row[1][:-1]))
```

```python
#Task 3.2

def merge(arr, low, mid, high):
    left = arr[low:mid+1]
    right = arr[mid+1:high+1]
    l = 0
    r = 0
    a = low
    while l < len(left) and r < len(right):
        if left[l].get_birthdate() <= right[r].get_birthdate():
            #append left when it is smaller
            arr[a] = left[l]
            l += 1
            a += 1
        else:
            arr[a] = right[r]
            r += 1
            a += 1
    #fill leftovers
    while l < len(left):
        arr[a] = left[l]
        l += 1
        a += 1
    while r < len(right):
        arr[a] = right[r]
        r += 1
        a += 1

def mergesort(arr, low, high):
    if low < high:
        mid = (low + high) // 2
        mergesort(arr, low, mid)
        mergesort(arr, mid + 1, high)
        merge(arr, low, mid, high)
    return arr

def sort_birthdays(birthdays):
    return mergesort(birthdays, 0, len(birthdays)-1)

sorted_birthdays = sort_birthdays(birthdays)
sorted_birthdays
```

```python
#Task 3.4

def binarysearch(arr, low, high, data):
    length = high
    while low <= high:
        mid = (low + high) // 2
        if data == arr[mid].get_birthdate():
            low = mid
            high = mid
            while low >= 0 and arr[low].get_birthdate() == data:
                low -= 1
            while high <= length and arr[high].get_birthdate() == data:
                high += 1
            return (low + 1, high - 1)
        elif data < arr[mid].get_birthdate():
            high = mid - 1
        else:
            low = mid + 1
    return (-1,-1)

def search_birthday(sorted_birthdays, date):
    return binarysearch(sorted_birthdays, 0, len(sorted_birthdays)-1, str(date))

search_birthday(sorted_birthdays, '19980701')
# test = [1,1,1,1,1,2,3,5,6,8,8,6,666,6,666,666,6,5,4,3,2,6,4]
# test.sort()
# print(test)
# binarysearch(test,0,len(test)-1,666)
```

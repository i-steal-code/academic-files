```python
#Task 2.1

def task2_1(filename):
    with open(filename) as file:
        output = []
        for row in file:
            row = row.strip()
            output.append(int(row))
    return output

result = task2_1('TEN.txt')
print(result)
print(len(result))
```

```python
#Task 2.2

def task2_2(list_of_integers):
    output = [_ for _ in list_of_integers]
    for n in range(1,len(output)):
        current = output[n]
        i = n - 1
        while i >= 0 and output[i] > current:
            output[i+1] = output[i]
            i -= 1
        output[i+1] = current
    return output

print(task2_2(task2_1('TEN.txt')))
```

```python
#Task 2.3

def partition(arr, first, last):
    pivot = arr[first]
    left = first + 1
    right = last 
    done = False
    while not done:
        while left <= right and arr[left] <= pivot:
            left += 1
        while right >= left and arr[right] >= pivot:
            right -= 1
        if left > right:
            done = True
        else:
            temp = arr[right]
            arr[right] = arr[left]
            arr[left] = temp
    arr[first] = arr[right]
    arr[right] = pivot
    return right
    
def quicksort(arr, first, last):
    if first < last:
        split = partition(arr, first, last)
        quicksort(arr, first, split-1)
        quicksort(arr, split+1, last)
    return arr

def task2_3(list_of_integers):
    return quicksort(list_of_integers, 0, len(list_of_integers)-1)

print(task2_3(task2_1('TEN.txt')))
```

```python
#Task 2.4

import timeit
```

```python
#comparing with t10

t10 = task2_1('TEN.txt')
print("insertion sort: ", [timeit.timeit(lambda: task2_2(t10), number=1) for _ in range(5)])
print("quick sort: ", [timeit.timeit(lambda: task2_3(t10), number=1) for _ in range(5)])
```

```python
#comparing with t100

t100 = task2_1('HUNDRED.txt')
print("insertion sort: ", [timeit.timeit(lambda: task2_2(t100), number=1) for _ in range(5)])
print("quick sort: ", [timeit.timeit(lambda: task2_3(t100), number=1) for _ in range(5)])
```

```python
#comparing with t1000

t1000 = task2_1('THOUSAND.txt')
print("insertion sort: ", [timeit.timeit(lambda: task2_2(t1000), number=1) for _ in range(5)])
print("quick sort: ", [timeit.timeit(lambda: task2_3(t1000), number=1) for _ in range(5)])
```

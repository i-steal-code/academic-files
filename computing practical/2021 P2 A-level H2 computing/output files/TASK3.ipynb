```python
#Task 3.1

class node:
    def __init__(self, data = None):
        self.__data = data
        self.__next = None
        
    def get_next(self):
        return self.__next
    
    def get_data(self):
        return self.__data
    
    def set_next(self, next):
        self.__next = next
        
    def set_data(self, data):
        self.__data = data
        
class LinkedList:
    def __init__(self):
        self.__head = None
        
    def get_head(self):
        return self.__head
    
    def set_head(self, new):
        self.__head = new
        
    def is_empty(self):
        return self.__head is None
        
    def insert(self, integer_value): #insert at head of list
        new = node(integer_value)
        #empty head case
        if self.is_empty():
            self.__head = new
            return True
        #regular case
        new.set_next(self.__head)
        self.__head = new
        return True
        
    def delete(self, integer_value):
        #empty head case
        if self.is_empty():
            return None
        #delete head case
        elif self.__head.get_data() == integer_value:
            self.__head = self.__head.get_next()
            return True
        #regular cases
        previous = None
        current = self.__head
        while current is not None and current.get_data() != integer_value:
            previous = current
            current = current.get_next()
        previous.set_next(current.get_next())
        
    def search(self, integer_value):
        #empty head case
        if self.is_empty():
            return None
        #regular cases
        current = self.__head
        while current is not None and current.get_data() != integer_value:
            current = current.get_next()
        return bool(current)
    
    def count(self):
        output = 0
        current = self.__head
        while current is not None:
            current = current.get_next()
            output += 1
        return output
    
    def to_string(self):
        output = []
        current = self.__head
        while current is not None:
            output.append(current.get_data())
            current = current.get_next()
        return str(output)
    
test = LinkedList()

with open('Task3data.txt') as file:
    for row in file:
        row = int(row.strip())
        test.insert(row)
    print(test.to_string())
    print('count: ',test.count())
    test.delete(8)
    print(test.to_string())
    print(test.search(7))
        
```

```python
#Task 3.2

class SortedLinkedList(LinkedList):
    def __init__(self):
        super().__init__()
        
    def insert(self, integer_value):
        new = node(integer_value)
        #empty head case
        if self.get_head() is None:
            self.set_head(new)
            return True
        #insert head case
        elif self.get_head().get_data() > integer_value:
            new.set_next(self.get_head())
            self.set_head(new)
            return True
        #regular cases
        previous = None
        current = self.get_head()
        while current is not None and current.get_data() < integer_value:
            previous = current
            current = current.get_next()
        new.set_next(current)
        previous.set_next(new)
        return True

test = SortedLinkedList()
with open('Task3data.txt') as file:
    for row in file:
        row = int(row.strip())
        test.insert(row)
    print(test.to_string())
    print('count: ',test.count())
    test.delete(8)
    print(test.to_string())
    print(test.search(94))
```

```python
#Task 3.3

class Queue(LinkedList):
    def __init__(self):
        super().__init__()
        
    def enqueue(self, integer_value):
        new = node(integer_value)
        #empty case
        if self.get_head() is None:
            self.set_head(new)
            return True
        #regular case
        previous = None
        current = self.get_head()
        while current is not None:
            previous = current
            current = current.get_next()
        #current is now none and previous is last node
        previous.set_next(new)
        return True
    
    def dequeue(self):
        #empty queue
        if self.get_head() is None:
            return None
        #dequeue head
        head = self.get_head()
        newhead = head.get_next()
        self.set_head(newhead)
        return head.get_data()
    
test = Queue()
with open('Task3data.txt') as file:
    for row in file:
        row = int(row.strip())
        test.enqueue(row)
    print(test.to_string())
    print("count: ",test.count())
    for i in range(5):
        print(test.dequeue())
    print(test.to_string())
```

```python
#Task 1.1

def task1_1(input_value):
    value = str(input_value)
    if not value.isnumeric():
#         print('invalid input')
        return -1
    if len(value) != 14:
#         print('wrong input length')
        return -1
    #split everything up
    value = [_ for _ in value]
    #mulitiply every 2 items
    for n in range(7):
        value[2*n+1] = str(int(value[2*n+1])*2)
    #re-coalesce everything
    collector = ''
    for i in value:
        collector += i
    #re-split
    value = [_ for _ in collector]
    #sum
    output = 0
    for i in value:
        output += int(i)
    return 10 - (output % 10)

task1_1(14576567654934)
```

```python
#Task 1.2
print(task1_1('16273849502536') == 2)
print(task1_1(16273849502536) == 2)
print(task1_1('ji') == -1)
print(task1_1('38288') == -1)
```

```python
#Task 1.3

def task1_3(input_value):
    value = str(input_value)
    if not value.isnumeric():
#         print('invalid input')
        return False
    if len(value) != 15:
#         print('wrong input length')
        return False
    check = value[-1]
    value = value[:-1]
    correct = task1_1(value)
    return int(check) == int(correct)

print(task1_3('162738495025362'))
print(task1_3(162738495025362))
print(not task1_3('ji'))
print(not task1_3('38288'))
```
